## ماشین بردار پشتیبان (SVM)

این روزها عبور از تنگهٔ هرمز برای کشتی‌های تجاری فقط یک خط روی نقشه نیست؛ هر تصمیم اشتباه می‌تواند جان خدمه، محموله و برنامهٔ چند بندر را به خطر بیندازد. در این سناریوی خیالی، مرکز هماهنگی دریایی بندرعباس برای هدایت کشتی‌هایی که منتظر عبورند، از یک چراغ اضطراری روی جزیرهٔ لارک استفاده می‌کند؛ چیزی شبیه چراغ راهنمایی، فقط برای نفتکش‌های چندصد‌متری.

وقتی ارتباط رادیویی شلوغ یا نامطمئن است، چراغ یکی از دو پیام «مسیر باز است؛ عبور کنید» یا «فعلاً در منطقهٔ انتظار بمانید» را با الگویی تکرارشونده از تغییرات روشنایی می‌فرستد. دوربین هر کشتی چند لحظه از این چراغ فیلم می‌گیرد و سامانه آن را به یک دنبالهٔ عددی تبدیل می‌کند. ناخداها ترجیح می‌دهند این بار مدل واقعاً چراغ را درست بخواند، نه اینکه فقط با اعتمادبه‌نفس زیاد پاسخ اشتباه بدهد.

سامانه پیش‌تر با نمونه‌های برچسب‌دار آزمایش شده است. وظیفهٔ شما این است که با استفاده از آن نمونه‌ها، پیام مربوط به کشتی‌های منتظر عبور را تشخیص دهید.

## داده‌ها

هر نمونه شامل **۲۴ اندازه‌گیری متوالی با فاصلهٔ زمانی برابر** است که دقیقاً یک چرخهٔ کامل از پیام را پوشش می‌دهند. مؤلفه‌های بردار به ترتیب زمان قرار گرفته‌اند:

</div>

$$
x=(x_0,x_1,\ldots,x_{23})\in\mathbb{R}^{24}.
$$

<div dir="rtl" style="text-align: right;">

| فایل | آرایه | ابعاد | توضیح |
| --- | --- | --- | --- |
| `train_data.npz` | `X` | `(240, 24)` | نمونه‌های آموزش |
| `train_data.npz` | `y` | `(240,)` | برچسب آموزش؛ ۱۲۰ نمونه از هر کلاس |
| `test_data.npz` | `X` | `(800, 24)` | نمونه‌های آزمون، بدون برچسب |
| `test_data.npz` | `id` | `(800,)` | شناسهٔ هر نمونه برای تطبیق خروجی |

برچسب `0` به معنای «مسیر باز است» و برچسب `1` به معنای «منتظر بمانید» است. آرایهٔ `id` شناسه است و برچسب کلاس را نشان نمی‌دهد.

## نمونه‌ها چگونه ساخته شده‌اند؟

- چراغ پیام خود را به‌صورت یک چرخهٔ پیوسته تکرار می‌کند. هر نمونه ۲۴ عدد دارد و دقیقاً یک دور کامل از این چرخه را نشان می‌دهد.
- دوربین ممکن است فیلم‌برداری را از بخش متفاوتی از چرخه آغاز کند. در داده‌های آموزش، هر نمونه می‌تواند یک یا دو پالس زودتر یا دیرتر شروع شود. در داده‌های آزمون، تفاوت زمان شروع ممکن است بیشتر باشد.
- با وجود جابه‌جایی نقطهٔ شروع، جهت زمان همیشه ثابت است: عددها به همان ترتیبی آمده‌اند که دوربین دیده است و فهرست ۲۴تایی هیچ‌وقت از آخر به اول برگردانده نشده است.
- فاصلهٔ کشتی تا چراغ و تنظیمات دوربین می‌تواند کل یک نمونه را روشن‌تر یا کم‌نورتر کند. نور محیط هم ممکن است یک مقدار ثابت به همهٔ ۲۴ عدد همان نمونه اضافه کند. این تغییرها درون یک نمونه ثابت‌اند، ولی از یک نمونه تا نمونهٔ دیگر فرق می‌کنند. اگر نویز را نادیده بگیریم و دو نمونه از یک نقطهٔ چرخه شروع شده باشند، رابطهٔ آن‌ها چنین است:

</div>

$$
x'=a x+b\mathbf{1},\qquad a>0,\quad b\in\mathbb{R}.
$$

<div dir="rtl" style="text-align: right;">

در این رابطه، `a` تغییر شدت نور و `b` اثر نور زمینه را نشان می‌دهد و بردار تمام‌یک ۲۴ مؤلفه دارد. مقدار `a` و `b` برای هیچ نمونه‌ای در اختیار شما نیست.

- نمونه‌ها کمی نویز دارند و حتی نمونه‌های متعلق به یک پیام کاملاً یکسان نیستند.
- دادهٔ گمشده، اشباع حسگر یا بریده‌شدن مقادیر وجود ندارد.
- داده‌ها شبیه‌سازی شده‌اند؛ دانش مخابرات یا دریانوردی برای حل سؤال لازم نیست.

```python
from pathlib import Path
from zipfile import ZIP_DEFLATED, ZipFile

NOTEBOOK_FILENAME = "X_SVM_QUESTIONS.ipynb"
CSV_FILENAME = "submission.csv"
ZIP_FILENAME = "submission.zip"

test_pred = np.asarray(fit_predict(X, y, X_test))
if test_pred.ndim != 1:
    raise ValueError("Predictions must be one-dimensional.")
if len(test_pred) != len(X_test) or len(test_pred) != len(test_ids):
    raise ValueError("Prediction count does not match test row count.")
if pd.isna(test_pred).any():
    raise ValueError("Predictions contain invalid values.")
if not np.isin(test_pred, [0, 1]).all():
    raise ValueError("Labels must be 0 or 1.")

notebook_path = Path(NOTEBOOK_FILENAME)
if not notebook_path.is_file():
    raise FileNotFoundError(
        f"Save this notebook as {NOTEBOOK_FILENAME} in the current folder, then run this cell again."
    )

submission = pd.DataFrame({
    "id": test_ids,
    "label": test_pred.astype(int),
})
if submission["id"].isna().any() or submission["id"].duplicated().any():
    raise ValueError("Identifier values must be present and unique.")

submission.to_csv(CSV_FILENAME, index=False)
with ZipFile(ZIP_FILENAME, "w", compression=ZIP_DEFLATED) as archive:
    archive.write(CSV_FILENAME, arcname=CSV_FILENAME)
    archive.write(notebook_path, arcname=NOTEBOOK_FILENAME)

print(f"{ZIP_FILENAME} created successfully.")
display(submission.head())
```
