# بازگشت خط نورد فولاد به مدار پس از محدودیت انرژی

خطوط نورد در یک مجتمع فولادی برای تولید پیوسته طراحی شده‌اند. با این حال، در دوره‌های اوج مصرف برق ممکن است کارخانه ناچار شود بار الکتریکی خط را کاهش دهد یا بخشی از تجهیزات را برای مدتی از مدار خارج کند.

بازگرداندن خط به وضعیت تولید تنها با روشن‌کردن دوبارهٔ تجهیزات انجام نمی‌شود. کوره، موتورهای اصلی، غلتک‌ها، سامانه‌های هیدرولیک و واحد خنک‌کاری باید با ترتیب مشخصی فعال شوند و پیش از ورود شمش به خط به شرایط مناسب برسند. تشخیص اشتباه وضعیت خط می‌تواند باعث افزایش مصرف لحظه‌ای برق، توقف دوبارهٔ تجهیزات، آسیب به محصول یا فرسودگی اجزای مکانیکی شود.

برای پایش این فرایند، ۲۰ سیگنال از بخش‌های مختلف خط ثبت می‌شود. کارشناسان اتاق کنترل، هر رکورد را با توجه به وضعیت واقعی خط در یکی از چهار مرحله قرار داده‌اند:

| برچسب | وضعیت عملیاتی | توضیح |
|---:|---|---|
| `0` | آماده‌باش و توقف کنترل‌شده | تجهیزات اصلی هنوز وارد مدار تولید نشده‌اند و خط در وضعیت ایمن نگه داشته شده است. |
| `1` | بازیابی حرارتی | بخش‌های حرارتی خط در حال رسیدن به شرایط لازم برای شروع فرایند هستند. |
| `2` | همگام‌سازی تجهیزات | موتورها، غلتک‌ها، سامانه‌های مکانیکی و مدارهای کنترلی برای آغاز تولید هماهنگ می‌شوند. |
| `3` | تولید زیر بار | خط وارد مرحلهٔ تولید شده و تجهیزات اصلی تحت بار عملیاتی قرار گرفته‌اند. |

هدف سامانه این است که با استفاده از اطلاعات ثبت‌شده، وضعیت فعلی خط را تشخیص دهد. این تشخیص به اپراتورها کمک می‌کند دربارهٔ ادامهٔ راه‌اندازی، ثابت نگه‌داشتن شرایط یا متوقف‌کردن فرایند تصمیم مناسب‌تری بگیرند.

وظیفهٔ شما ساخت یک مدل **MLP** است که برای هر رکورد موجود در `test.csv` یکی از چهار وضعیت `0`، `1`، `2` یا `3` را پیش‌بینی کند. ملاک داوری، کیفیت پیش‌بینی نهایی روی داده‌های آزمون است.

| Accuracy | نمره از ۱۰۰ |
|---:|---:|
| ۵۰٪ یا کمتر | ۰ |
| ۶۰٪ | ۲۵ |
| ۷۰٪ | ۷۵ |
| ۷۵٪ یا بیشتر | ۱۰۰ |

بین دو نقطهٔ متوالی، نمره به‌صورت خطی محاسبه می‌شود. برای مثال، دقت ۶۵٪ نمرهٔ ۵۰ و دقت ۷۲٪ نمرهٔ ۸۵ می‌گیرد.

</div>

$$
\mathrm{Score}(A)=
\begin{cases}
0 & A\le 0.50\\
250(A-0.50) & 0.50<A\le 0.60\\
25+500(A-0.60) & 0.60<A\le 0.70\\
75+500(A-0.70) & 0.70<A<0.75\\
100 & A\ge 0.75
\end{cases}

## ساخت فایل نهایی تحویل

پس از تکمیل راه‌حل، نوت‌بوک را با نام `sensor_fault_exam.ipynb` ذخیره کنید و سپس سلول نهایی زیر را اجرا کنید.

پیش از ارسال بررسی کنید که:

- فایل ZIP شامل `submission.csv` و `sensor_fault_exam.ipynb` باشد.
- نوت‌بوک داخل ZIP همان نسخهٔ تکمیل‌شده و ذخیره‌شدهٔ شما باشد.
- همهٔ `row_id`های آزمون دقیقاً یک‌بار در CSV آمده باشند.
- مقدارهای ستون `label` فقط ۰، ۱، ۲ یا ۳ باشند.

</div>

```python
from pathlib import Path
from zipfile import ZIP_DEFLATED, ZipFile

NOTEBOOK_FILENAME = "sensor_fault_exam.ipynb"
CSV_FILENAME = "submission.csv"
ZIP_FILENAME = "submission.zip"

if test_predictions is None:
    raise ValueError("test_predictions is still None.")

test_predictions = np.asarray(test_predictions)
if test_predictions.ndim != 1:
    raise ValueError("test_predictions must be one-dimensional.")
if len(test_predictions) != len(test):
    raise ValueError("Prediction count does not match test row count.")
if pd.isna(test_predictions).any():
    raise ValueError("Predictions contain missing values.")
if not np.isin(test_predictions, [0, 1, 2, 3]).all():
    raise ValueError("Every prediction must be one of 0, 1, 2, 3.")

notebook_path = Path(NOTEBOOK_FILENAME)
if not notebook_path.is_file():
    raise FileNotFoundError(
        f"Save this notebook as {NOTEBOOK_FILENAME} in the current folder, then run this cell again."
    )

submission = pd.DataFrame({
    "row_id": test["row_id"].to_numpy(),
    "label": test_predictions.astype(int),
})
if submission["row_id"].isna().any() or submission["row_id"].duplicated().any():
    raise ValueError("row_id values must be present and unique.")

submission.to_csv(CSV_FILENAME, index=False)
with ZipFile(ZIP_FILENAME, "w", compression=ZIP_DEFLATED) as archive:
    archive.write(CSV_FILENAME, arcname=CSV_FILENAME)
    archive.write(notebook_path, arcname=NOTEBOOK_FILENAME)

print(f"{ZIP_FILENAME} created successfully.")
display(submission.head())
```
