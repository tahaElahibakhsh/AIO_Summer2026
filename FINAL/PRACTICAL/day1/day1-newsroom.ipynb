## صورت مسئله

شما تازه به‌عنوان تنها مهندس دادهٔ یک آرشیو خبری کوچک استخدام شده‌اید. این آرشیو هزاران خبر متنی قدیمی (به زبان انگلیسی، خبرگزاری رویترز، دههٔ ۹۰ میلادی) دارد و سردبیر می‌خواهد بتوان در آرشیو بر اساس نام افراد، سازمان‌ها و مکان‌های ذکرشده جست‌وجو کرد. سردبیر می‌خواهد ببیند با امکانات مختلف (از یک قانون سادهٔ دستی گرفته تا یک مدل زبانی) چقدر می‌توان کیفیت این جست‌وجو را بالا برد.

باید یک سیستم **تشخیص موجودیت‌های نام‌دار (Named Entity Recognition / NER)** بسازید: برای هر کلمه در متن مشخص کنید که آیا بخشی از نام یک شخص (PER)، سازمان (ORG)، مکان (LOC)، یا یک موجودیت دیگر (MISC) است یا نه (O).

این سؤال **کاملاً باز** است: دو تلاش اولیه (baseline) از قبل برایتان پیاده‌سازی شده — کافی است آن‌ها را اجرا کنید. هر دو روی **کل** دادهٔ آموزشی و آزمون اجرا می‌شوند و خروجی هرکدام را می‌توانید مستقیماً به فایل ارسالی تبدیل کنید. از همان‌جا به بعد، مسیر با خودتان است: می‌توانید baseline ها را بهبود دهید، جهت‌های دیگر را امتحان کنید، یا کار خودتان را از صفر بسازید.

## شناسنامه داده

داده از مجموعهٔ معروف و واقعی **CoNLL-2003** گرفته شده — یکی از استانداردترین بنچمارک‌های NER در دنیای NLP، برگرفته از خبرهای واقعی خبرگزاری رویترز در سال ۱۹۹۶. یعنی متن‌ها ساختگی نیستند: جمله‌های واقعی خبری با اسم افراد، تیم‌های ورزشی، کشورها و سازمان‌های واقعی آن دوران.

دو فایل در اختیار دارید: `train.csv` (با برچسب) و `test.csv` (بدون برچسب). هر سطر یک **کلمه (توکن)** است، نه یک جملهٔ کامل:

| ستون | توضیح |
|---|---|
| `sentence_id` | شناسهٔ جمله‌ای که این کلمه در آن قرار دارد (کلمات یک جمله را با هم گروه‌بندی کنید) |
| `token_idx` | موقعیت کلمه درون جمله (از صفر شروع می‌شود) |
| `token` | خود کلمه، دقیقاً همان‌طور که در متن اصلی آمده (با حروف بزرگ/کوچک اصلی — این خودش یک سرنخ مهم برای NER است) |
| `tag` | برچسب صحیح این کلمه (فقط در `train.csv`) |

برچسب‌ها به‌فرمت **BIO** هستند: `B-` یعنی شروع یک موجودیت، `I-` یعنی ادامهٔ همان موجودیت، و `O` یعنی خارج از هر موجودیتی. برای مثال «Peter Blackburn» به‌صورت `B-PER`, `I-PER` برچسب می‌خورد. چهار نوع موجودیت داریم: `PER` (شخص)، `ORG` (سازمان)، `LOC` (مکان)، `MISC` (سایر، مثل ملیت‌ها و رویدادها) — پس ۹ برچسب ممکن داریم: `O` و هشت ترکیب `B-`/`I-` با این چهار نوع.

یک نکتهٔ مهم دربارهٔ توزیع داده: حدود ۸۳٪ کلمات برچسب `O` دارند (کلمات معمولی، نه بخشی از یک نام). این یعنی **دقت سطح‌کلمه (accuracy) معیار گمراه‌کننده‌ای است** — مدلی که همیشه `O` حدس بزند هم دقت بالایی می‌گیرد، بدون اینکه هیچ موجودیتی را واقعاً پیدا کرده باشد. به همین دلیل معیار امتیازدهی این سؤال F1 در سطح موجودیت است، نه دقت.

فایل `test.csv` همان ستون‌های `train.csv` را دارد، به‌جز ستون `tag`. برچسب‌های آن پنهان است و نمرهٔ نهایی با آن‌ها محاسبه می‌شود.

## معیار امتیازدهی

F1 **در سطح موجودیت** (بازه‌ی دقیق موجودیت و نوع موجودیت) — یک پیش‌بینی فقط زمانی درست حساب می‌شود که کل بازهٔ کلمات و نوع موجودیت **دقیقاً** با پاسخ صحیح یکسان باشد.

| | F1 |
|---|---|
| baseline (بهترین دو تلاش داده‌شده) | ۰٫۱۲۰ |
| full score | ۰٫۷۸۲ |

نمرهٔ نرمال‌شده (عددی بین ۰ تا ۱۰۰) این‌طور محاسبه می‌شود:

</div>

$$
\text{score} = 100 \times \operatorname{clip}\!\left(\frac{F_1 - F_1^{\text{baseline}}}{F_1^{\text{full}} - F_1^{\text{baseline}}},\; 0,\; 1\right), \qquad F_1^{\text{baseline}} = 0.120,\quad F_1^{\text{full}} = 0.782
$$

<div dir="rtl" style="text-align: right; font-family: Tahoma, 'IRANSans', 'Vazirmatn', 'Segoe UI', sans-serif;">

یعنی F1 برابر با baseline نمرهٔ ۰ و F1 برابر با full score (یا بیشتر) نمرهٔ ۱۰۰ می‌گیرد و بین این دو نمره به‌طور خطی زیاد می‌شود؛ F1 کمتر از baseline هم نمرهٔ صفر دارد.

</div>

## بارگذاری داده

</div>

```python
import numpy as np
import pandas as pd

# keep_default_na=False: tokens such as "NA" or "null" are real words here, not missing values
train = pd.read_csv("train.csv", keep_default_na=False)
test = pd.read_csv("test.csv", keep_default_na=False)
print("train:", train.shape, " test:", test.shape)
train.head(10)
```

```python
# Reshape the flat token-per-row table into a list of sentences, since
# almost every method below wants (tokens, tags) per sentence.
def to_sentences(df):
    out = []
    for sid, g in df.sort_values(["sentence_id", "token_idx"]).groupby("sentence_id", sort=True):
        out.append({
            "sentence_id": sid,
            "token_idx": list(g.token_idx),
            "tokens": list(g.token),
            "tags": list(g.tag) if "tag" in g.columns else None,
        })
    return out


def to_row_predictions(df, sents, sent_tags):
    """Flatten per-sentence tag lists back into one tag per row of df, in df's row order."""
    lookup = {}
    for s, tags in zip(sents, sent_tags):
        if len(tags) != len(s["tokens"]):
            raise ValueError(f"sentence {s['sentence_id']}: {len(tags)} tags for {len(s['tokens'])} tokens")
        for idx, tag in zip(s["token_idx"], tags):
            lookup[(s["sentence_id"], idx)] = tag
    return [lookup[(sid, idx)] for sid, idx in zip(df.sentence_id, df.token_idx)]


train_sents = to_sentences(train)
test_sents = to_sentences(test)
print(f"{len(train_sents)} train sentences, {len(test_sents)} test sentences")
train_sents[0]
```

## معیار F1 (سطح موجودیت)

در این معیار هر موجودیت (مثلاً «Peter Blackburn» از نوع `PER`) یک واحد حساب می‌شود، نه هر کلمه: **precision** یعنی چه کسری از موجودیت‌هایی که مدل پیدا کرده کاملاً درست‌اند، **recall** یعنی چه کسری از موجودیت‌های واقعی را مدل کاملاً درست پیدا کرده، و **F1** میانگین هارمونیک این دو است. اگر فقط یکی از کلمات یک موجودیت یا نوع آن اشتباه باشد، آن موجودیت هم در precision و هم در recall غلط حساب می‌شود.

این تابع را برای اندازه‌گیری همهٔ تلاش‌هایتان استفاده کنید. چون برچسب‌های `test.csv` پنهان است، سلول بعدی بخشی از جمله‌های `train` را به‌عنوان دادهٔ اعتبارسنجی (`val_sents`) کنار می‌گذارد؛ روی `fit_sents` آموزش دهید و روی `val_sents` بسنجید.

</div>

```python
# Entity-level (exact span + exact type) F1 -- a prediction only counts as
# correct if the ENTIRE span boundaries and type match the gold entity.
def entity_spans(tags):
    out = set()
    i = 0
    while i < len(tags):
        if tags[i].startswith("B-"):
            etype = tags[i][2:]
            j = i + 1
            while j < len(tags) and tags[j] == f"I-{etype}":
                j += 1
            out.add((i, j, etype))
            i = j
        else:
            i += 1
    return out

def entity_f1(y_true, y_pred):
    tp = fp = fn = 0
    for t, p in zip(y_true, y_pred):
        ts, ps = entity_spans(t), entity_spans(p)
        tp += len(ts & ps)
        fp += len(ps - ts)
        fn += len(ts - ps)
    prec = tp / (tp + fp) if (tp + fp) else 0.0
    rec = tp / (tp + fn) if (tp + fn) else 0.0
    f1 = 2 * prec * rec / (prec + rec) if (prec + rec) else 0.0
    return prec, rec, f1
```

```python
# Local validation split: the test labels are hidden, so hold out 20% of the
# training sentences. Fit on fit_sents, measure on val_sents.
rng = np.random.RandomState(0)
perm = rng.permutation(len(train_sents))
n_val = len(train_sents) // 5
val_sents = [train_sents[i] for i in perm[:n_val]]
fit_sents = [train_sents[i] for i in perm[n_val:]]
print(f"fit: {len(fit_sents)} sentences, val: {len(val_sents)} sentences")
```

## Baseline ۱: قانون سادهٔ دستی

یک تابع ساده که چند لیست دستی + قانون «حرف اول بزرگ» را به‌کار می‌برد. بدون یادگیری، بدون داده — فقط چند خط قانون. چون این روش چیزی یاد نمی‌گیرد، آن را روی **همهٔ** جمله‌های `train` می‌سنجیم و بعد روی **همهٔ** جمله‌های `test` پیش‌بینی می‌کنیم (`rule_test_preds`).

</div>

```python
TITLES = {"mr.", "mrs.", "dr.", "president", "minister"}
ORG_SUFFIX = {"corp", "inc", "ltd", "co"}
MISC_WORDS = {"german", "british", "french", "english", "dutch", "american"}

def rule_baseline(tokens):
    tags = ["O"] * len(tokens)
    for i, tok in enumerate(tokens):
        low = tok.lower().strip(".,")
        if low in MISC_WORDS:
            tags[i] = "B-MISC"
        elif low in ORG_SUFFIX:
            tags[i] = "B-ORG"
        elif i > 0 and tokens[i - 1].lower() in TITLES and tok[:1].isupper():
            tags[i] = "B-PER"
        elif tok[:1].isupper() and tok.isalpha() and i != 0:
            tags[i] = "B-PER"
    return tags

# Self-check on ALL training sentences (the rule learns nothing, so no split is needed)
y_true = [s["tags"] for s in train_sents]
y_pred = [rule_baseline(s["tokens"]) for s in train_sents]
p, r, f1 = entity_f1(y_true, y_pred)
print(f"rule-based baseline on all of train (n={len(train_sents)}): P={p:.3f} R={r:.3f} F1={f1:.3f}")

# Predict every test sentence -> one tag per row of test.csv, ready for submission
rule_test_preds = to_row_predictions(test, test_sents, [rule_baseline(s["tokens"]) for s in test_sents])
print(f"rule_test_preds: {len(rule_test_preds)} tags for {len(test)} test rows")
```

```python
# Full submission-format CSV for baseline 1: one row per row of test.csv (sentence_id, token_idx, tag).
rule_submission = test[["sentence_id", "token_idx"]].copy()
rule_submission["tag"] = rule_test_preds
rule_submission.to_csv("rule_baseline_submission.csv", index=False)
print(f"rule_baseline_submission.csv: {len(rule_submission)} rows for {len(test)} test rows")
rule_submission.head()
```

## Baseline ۲: LLM با zero-shot prompting

از یک مدل زبانی کوچک محلی (`Qwen3-0.6B`، چون این‌جا نه GPU داریم نه دسترسی به اینترنت — وزن‌هایش از قبل روی سیستم شما دانلود شده) با یک **prompt قابل‌تنظیم** می‌خواهیم موجودیت‌ها را استخراج کند — بدون هیچ نمونهٔ آموزشی در prompt. متغیر `PROMPT_TEMPLATE` را هر طور دوست دارید عوض کنید و دوباره اجرا کنید تا ببینید امتیاز چطور تغییر می‌کند.

</div>

```python
import os
import time
os.environ["HF_HUB_OFFLINE"] = "1"  # only the pre-downloaded checkpoints below are available -- no network
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

# The weights are shipped as plain local folders inside model_weights/, so
# from_pretrained() below reads straight from disk -- zero internet needed.
MODEL_NAME = "model_weights/Qwen3-0.6B"
VALID_TYPES = {"PER", "ORG", "LOC", "MISC"}

# <<< CHANGE THIS to experiment with different instructions >>>
PROMPT_TEMPLATE = (
    "You are a precise named-entity tagger. Given a sentence, output each "
    "entity you find as 'entity_text : TYPE' on its own line, where TYPE is "
    "one of PER, ORG, LOC, MISC. If there are no entities, output 'none'."
)

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
llm = AutoModelForCausalLM.from_pretrained(MODEL_NAME)
llm.eval()

def parse_llm_answer(text, tokens):
    # Turn the model's free-text answer ("entity : TYPE" lines) back into
    # BIO tags aligned with the original tokens.
    tags = ["O"] * len(tokens)
    low_tokens = [w.lower() for w in tokens]
    if not text or text.strip().lower() == "none":
        return tags
    chunks = [c for line in text.split("\n") for c in line.split("|")]
    for chunk in chunks:
        if ":" not in chunk:
            continue
        ent_str, etype = chunk.rsplit(":", 1)
        ent_str = ent_str.strip().strip("-").strip()
        etype = etype.strip().upper()
        if etype not in VALID_TYPES or not ent_str:
            continue
        ent_words = [w.lower() for w in ent_str.split()]
        L = len(ent_words)
        if L == 0:
            continue
        for i in range(len(tokens) - L + 1):
            if low_tokens[i:i + L] == ent_words and tags[i] == "O":
                tags[i] = f"B-{etype}"
                for k in range(1, L):
                    tags[i + k] = f"I-{etype}"
                break
    return tags

def llm_zeroshot_predict(tokens, prompt=PROMPT_TEMPLATE):
    sentence = " ".join(tokens)
    messages = [{"role": "system", "content": prompt},
                {"role": "user", "content": f"Sentence: {sentence}"}]
    # enable_thinking=False: Qwen3 models "think" out loud by default (a
    # <think>...</think> block before the real answer), which would eat the
    # whole generation budget below before any answer appears -- turn it off.
    text_in = tokenizer.apply_chat_template(messages, tokenize=False,
                                             add_generation_prompt=True, enable_thinking=False)
    inputs = tokenizer(text_in, return_tensors="pt", truncation=True, max_length=768)
    with torch.no_grad():
        out_ids = llm.generate(**inputs, max_new_tokens=60, do_sample=False,
                               pad_token_id=tokenizer.eos_token_id)
    gen = tokenizer.decode(out_ids[0][inputs["input_ids"].shape[1]:], skip_special_tokens=True)
    return parse_llm_answer(gen, tokens)

def llm_predict_all(token_lists, prompt=PROMPT_TEMPLATE, log_every=0):
    """Tag many sentences one by one, printing progress every `log_every` sentences."""
    results, t0 = [], time.time()
    for n, tokens in enumerate(token_lists, 1):
        results.append(llm_zeroshot_predict(tokens, prompt))
        if log_every and (n % log_every == 0 or n == len(token_lists)):
            print(f"  {n}/{len(token_lists)} sentences, {time.time() - t0:.0f}s elapsed")
    return results

# a quick look at 3 sentences
demo = train_sents[:3]
for ex in demo:
    pred = llm_zeroshot_predict(ex["tokens"])
    print(list(zip(ex["tokens"], pred, ex["tags"])))
```

```python
# Self-check on the validation sentences. 
N_LLM_CHECK = 100
check = val_sents if N_LLM_CHECK is None else val_sents[:N_LLM_CHECK]
y_true = [s["tags"] for s in check]
y_pred = llm_predict_all([s["tokens"] for s in check], log_every=25)
p, r, f1 = entity_f1(y_true, y_pred)
print(f"LLM zero-shot baseline (self-check, n={len(check)}): P={p:.3f} R={r:.3f} F1={f1:.3f}")
```

```python
# Predict ALL test sentences -> one tag per row of test.csv, ready for submission.
llm_test_sent_preds = llm_predict_all([s["tokens"] for s in test_sents], log_every=250)
llm_test_preds = to_row_predictions(test, test_sents, llm_test_sent_preds)
print(f"llm_test_preds: {len(llm_test_preds)} tags for {len(test)} test rows")
```

```python
# Full submission-format CSV for baseline 2: 
llm_submission = test[["sentence_id", "token_idx"]].copy()
llm_submission["tag"] = llm_test_preds
llm_submission.to_csv("llm_baseline_submission.csv", index=False)
print(f"llm_baseline_submission.csv: {len(llm_submission)} rows for {len(test)} test rows")
llm_submission.head()
```

همان‌طور که می‌بینید، هیچ‌کدام از این دو baseline امتیاز بالایی نمی‌گیرد (روی برچسب‌های پنهان `test`: قانون دستی F1=۰٫۰۵۶ و LLM با zero-shot، که کمی بهتر است، F1=۰٫۱۲۰ — همین عدد بزرگ‌تر، baseline رسمی سؤال است). این نقطهٔ شروع شماست، نه هدف.

فایل کامل پیش‌بینی هرکدام، با همان قالب `submission.csv` (ستون‌های `sentence_id`, `token_idx`, `tag` و یک سطر برای هر سطر `test.csv`)، هم زیر همان baseline در `rule_baseline_submission.csv` و `llm_baseline_submission.csv` ذخیره شده است. این دو فایل فقط برای بررسی و مقایسه‌اند؛ چیزی که بارگذاری می‌کنید همچنان `submission.zip` است که سلول آخر می‌سازد.

## قالب اختیاری: fine-tune کردن DistilBERT

اگر خواستید `distilbert-base-cased` را برای برچسب‌زنی کلمات fine-tune کنید، سلول‌های زیر یک قالب نیمه‌کامل است:

- **آماده‌سازی داده هم با خودتان است — دو تابع `TODO` داخل `finetune_bert_template`:**
  - `encode(tokens, tags)`: یک جمله (لیست کلمات) را با توکنایزر BERT به زیرکلمه‌ها (subword) می‌شکند (حداکثر ۹۶ زیرکلمه) و برای **هر زیرکلمه** دقیقاً یک برچسب می‌سازد. برچسب هر کلمه فقط روی **اولین** زیرکلمه‌اش گذاشته می‌شود؛ زیرکلمه‌های بعدی همان کلمه و توکن‌های ویژهٔ `[CLS]` و `[SEP]` برچسب `-100` می‌گیرند (`enc.word_ids()` می‌گوید هر زیرکلمه از کدام کلمه آمده). خروجی: سه لیست هم‌طول `(input_ids, attention_mask, labels)`.
  - `collate(batch)`: جمله‌های یک batch طول‌های متفاوتی دارند؛ این تابع همه را تا طول بلندترین جملهٔ همان batch پر (pad) می‌کند و به tensor تبدیل می‌کند: `input_ids` با شناسهٔ توکن `[PAD]` (`bert_tok.pad_token_id`)، `attention_mask` با `0` تا مدل به این موقعیت‌ها توجه نکند، و `labels` با `-100`. خروجی: سه tensor به شکل `(batch, max_len)`.
  - **ترفند `-100`:** تابع loss مدل (`CrossEntropyLoss` در PyTorch) به‌طور پیش‌فرض هر موقعیتی را که برچسبش `-100` باشد (`ignore_index=-100`) کاملاً نادیده می‌گیرد: نه در میانگین loss حساب می‌شود و نه گرادیانی تولید می‌کند. پس با گذاشتن `-100` روی زیرکلمه‌های غیر اول، توکن‌های ویژه و padding، مدل فقط روی «یک پیش‌بینی برای هر کلمه» آموزش می‌بیند — همان چیزی که بعداً ارزیابی می‌شود — بدون اینکه ماسک جداگانه‌ای برای loss بسازید.
  - `Dataset` (کلاس `SimpleDS`) و `DataLoader` (متغیر `dl`) آماده‌اند و از همین دو تابع استفاده می‌کنند؛ هر batch سه‌تایی `(input_ids, attention_mask, labels)` است.
- **ساخت مدل و حلقهٔ آموزش با خودتان است:** از خط `bert_model =` به بعد را کامل کنید تا تابع جفت `(bert_tok, bert_model)` آموزش‌دیده را برگرداند.
- **ارزیابی و ساخت فایل پیش‌بینی آماده است:** `predict_bert_template` خروجی مدل را به یک برچسب برای هر کلمه برمی‌گرداند و `evaluate_bert_template` امتیاز F1 سطح موجودیت را روی جمله‌های برچسب‌دار (مثلاً `val_sents`) چاپ می‌کند. سلول سوم، بعد از اینکه `RUN_BERT = True` کنید، مدل را آموزش می‌دهد، روی `val_sents` می‌سنجد، برای همهٔ سطرهای `test.csv` پیش‌بینی می‌کند (`bert_test_preds`) و آن را در `bert_predictions.csv` با قالب فایل ارسالی ذخیره می‌کند.
