# نقشه‌برداری اضطراری با پنج برچسب

## صورت مسئله

چند ساعت بعد از یک زلزله یا سیل بزرگ، تیم‌های امداد به نقشه نیاز دارند: فرودگاه‌ها و باندهایی که هواپیماهای امدادی می‌توانند روی آن‌ها بنشینند کجاست؟ کدام پل‌ها، روگذرها و ایستگاه‌های راه‌آهن مسیر رساندن کمک‌اند؟ مناطق مسکونی متراکم، که بیشترین آسیب‌دیده را دارند، کجا هستند؟ بندرها، مخزن‌های سوخت و نیروگاه‌ها کجا قرار دارند؟

تصاویر هوایی تازه‌ی کل منطقه زود می‌رسند؛ ده‌ها هزار قطعه تصویر. اما برچسب‌زدن آن‌ها کار انسان است: گروهی از داوطلبان نقشه‌برداری در فرصت کوتاهی که داشتند، برای هر نوع منطقه فقط **۵ قطعه** را با اطمینان برچسب زده‌اند. با ۴۵ نوع منطقه، که بعضی‌شان بسیار شبیه هم‌اند (مسکونی متراکم، متوسط و پراکنده؛ راه‌آهن و ایستگاه راه‌آهن؛ مزرعه‌ی دایره‌ای و مستطیلی)، برچسب‌زدن دستی بقیه‌ی تصاویر به‌موقع تمام نمی‌شود.

شما باید سامانه‌ای بسازید که نوع هر قطعه‌ی تصویر را تشخیص دهد.

## مدل‌های ازپیش‌آموزش‌دیده‌ی مجاز

وزن‌های سه مدل زیر در پوشه‌ی `weights/` کنار همین نوت‌بوک قرار دارد (سامانه‌ی داوری به اینترنت دسترسی ندارد). هر سه با کتابخانه‌ی `torchvision` ساخته می‌شوند و روی **ImageNet-1k** آموزش دیده‌اند: حدود ۱.۲۸ میلیون عکس معمولی در ۱۰۰۰ کلاس (حیوانات، اشیا، وسایل نقلیه، غذا و ...)، برای وظیفه‌ی **تشخیص شیء اصلی یک عکس**.

| مدل | فایل وزن | تعداد پارامتر | داده و وظیفه‌ی پیش‌آموزش | دقت top-1 روی ImageNet | اندازه‌ی بردار ویژگی |
|---|---|---|---|---|---|
| `mobilenet_v3_large` | `weights/mobilenet_v3_large.pth` | حدود ۵.۵ میلیون | ImageNet-1k، دسته‌بندی ۱۰۰۰ کلاسه | حدود ۷۵٪ | 960 |
| `efficientnet_b0` | `weights/efficientnet_b0.pth` | حدود ۵.۳ میلیون | ImageNet-1k، دسته‌بندی ۱۰۰۰ کلاسه | حدود ۷۸٪ | 1280 |
| `resnet18` | `weights/resnet18.pth` | حدود ۱۱.۷ میلیون | ImageNet-1k، دسته‌بندی ۱۰۰۰ کلاسه | حدود ۷۰٪ | 512 |

این مدل‌ها با تصاویر ۲۲۴×۲۲۴ و نرمال‌سازی ImageNet (میانگین `[0.485, 0.456, 0.406]` و انحراف معیار `[0.229, 0.224, 0.225]`) آموزش دیده‌اند. تابع `load_backbone` در پایین نوت‌بوک هر کدام را از روی فایل وزن می‌سازد.

### مسئله‌ی ما از جنس دیگری است

وظیفه‌ای که این مدل‌ها برایش آموزش دیده‌اند با وظیفه‌ی ما فرق دارد:

- **زاویه‌ی دید:** ImageNet عکس‌هایی است که انسان از روی زمین و از کنار گرفته؛ تصاویر ما عمود از بالا گرفته شده‌اند.
- **شیء در برابر صحنه:** در ImageNet یک شیء اصلی بیشتر قاب را پر می‌کند و برچسب، نام همان شیء است. برچسب‌های ما نوع یک **منطقه** است: الگوی خیابان‌ها، چیدمان ساختمان‌ها، شکل مزرعه‌ها. یک «فرودگاه» ترکیبی از باند، هواپیما و ساختمان است.
- **مقیاس:** بسته به تصویر، یک ساختمان ممکن است چند پیکسل یا نیمی از تصویر باشد.
- **جهت:** عکس‌های ImageNet «بالا» و «پایین» دارند؛ تصویر هوایی جهت مرجع ندارد.
- **کلاس‌ها:** بیشتر کلاس‌های ما (تالاب، بوته‌زار خشک، مسکونی پراکنده، زمین پلکانی ...) اصلاً در ImageNet وجود ندارند.

هیچ‌کدام از این مدل‌ها با تصویر هوایی برچسب‌دار آموزش ندیده‌اند.

- کد شما باید روی CPU هم بدون خطا اجرا شود تا بتوانید آن را روی رایانه‌ی خودتان امتحان کنید؛ روی CPU روش‌های سنگین بسیار کندترند، پس برای آزمایش محلی می‌توانید موقتاً بخشی از داده را استفاده کنید. زمان ۱۰ دقیقه روی GPU سامانه‌ی داوری سنجیده می‌شود.

## معیار ارزیابی و نمره

معیار، **macro-F1** روی ۴٬۵۰۰ تصویر آزمون است: F1 هر کلاس جداگانه حساب می‌شود و میانگین ساده‌ی ۴۵ کلاس گرفته می‌شود (تصاویر آزمون از هر کلاس ۱۰۰ تاست). نمره یک تابع خطیِ تکه‌ای از macro-F1 است:

</div>

$$
\text{score} =
\begin{cases}
0 & F1 \le 0.268 \\
30 \cdot \dfrac{F1 - 0.268}{0.629 - 0.268} & 0.268 < F1 \le 0.629 \\
30 + 70 \cdot \dfrac{F1 - 0.629}{0.793 - 0.629} & 0.629 < F1 < 0.793 \\
100 & F1 \ge 0.793
\end{cases}
$$

## بارگذاری داده و ابزارها

```python
import time
NOTEBOOK_T0 = time.time()   # the whole notebook must finish in < 10 minutes

import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.nn.functional as F
import torchvision.models as tvm

# The judging machine exposes exactly one GPU slice, always as "cuda:0".
# Without a GPU (e.g. a local laptop) everything falls back to the CPU.
# Put every model and tensor on DEVICE; never hard-code "cuda", "cuda:1", ".cuda()" etc.
DEVICE = "cuda:0" if torch.cuda.is_available() else "cpu"
USE_GPU = DEVICE.startswith("cuda")
if USE_GPU:
    torch.cuda.set_device(0)
print("device:", DEVICE, torch.cuda.get_device_name(0) if USE_GPU else "(no GPU: running on CPU)")

IMAGENET_MEAN = torch.tensor([0.485, 0.456, 0.406], device=DEVICE).view(1, 3, 1, 1)
IMAGENET_STD = torch.tensor([0.229, 0.224, 0.225], device=DEVICE).view(1, 3, 1, 1)


def to_tensor(X_uint8, size=128):
    """uint8 (N, H, W, 3) numpy -> ImageNet-normalized float tensor (N, 3, size, size) on DEVICE."""
    x = torch.from_numpy(X_uint8).to(DEVICE).permute(0, 3, 1, 2).float() / 255.0
    if size != x.shape[-1]:
        x = F.interpolate(x, size=size, mode="bilinear", align_corners=False)
    return (x - IMAGENET_MEAN) / IMAGENET_STD


def load_backbone(name):
    """Build one allowed ImageNet model from its local weight file.
    Returns (feature_extractor, feature_dim); the extractor maps images to one vector per image."""
    path = f"{WEIGHTS_DIR}/{name}.pth"
    if name == "mobilenet_v3_large":
        m = tvm.mobilenet_v3_large(weights=None); m.load_state_dict(torch.load(path, map_location="cpu"))
        return nn.Sequential(m.features, nn.AdaptiveAvgPool2d(1), nn.Flatten()).to(DEVICE), 960
    if name == "efficientnet_b0":
        m = tvm.efficientnet_b0(weights=None); m.load_state_dict(torch.load(path, map_location="cpu"))
        return nn.Sequential(m.features, nn.AdaptiveAvgPool2d(1), nn.Flatten()).to(DEVICE), 1280
    if name == "resnet18":
        m = tvm.resnet18(weights=None); m.load_state_dict(torch.load(path, map_location="cpu"))
        m.fc = nn.Identity()
        return m.to(DEVICE), 512
    raise ValueError(name)
```

```python
DATA_DIR = "data"
WEIGHTS_DIR = "weights"

lab = np.load(f"{DATA_DIR}/labeled.npz")
unl = np.load(f"{DATA_DIR}/unlabeled.npz")
tst = np.load(f"{DATA_DIR}/test.npz")

X_lab, y_lab, CLASSES = lab["X"], lab["y"].astype(np.int64), [str(c) for c in lab["classes"]]
X_unl = unl["X"]
X_test, test_ids = tst["X"], tst["id"]
N_CLASSES = len(CLASSES)

print("labeled  :", X_lab.shape, "| per class:", np.bincount(y_lab, minlength=N_CLASSES).min())
print("unlabeled:", X_unl.shape)
print("test     :", X_test.shape)
```

## مروری کوتاه بر روش‌های یادگیری با برچسب کم

وقتی داده‌ی برچسب‌دار بسیار کم و داده‌ی بدون برچسب فراوان است، این ایده‌ها رایج‌اند:

- **بازنمایی (embedding):** لایه‌ی دسته‌بند آخر یک مدل ازپیش‌آموزش‌دیده را حذف کنید تا برای هر تصویر یک بردار عددی بدهد. تصاویر مشابه باید بردارهای نزدیک به هم داشته باشند؛ معمولاً بردارها را نرمال می‌کنند و شباهت کسینوسی می‌گیرند.
- **k نزدیک‌ترین همسایه (kNN):** برای هر تصویر آزمون، k تصویر برچسب‌داری که بردارشان از همه شبیه‌تر است را پیدا کنید و بین برچسب‌هایشان رأی بگیرید (رأی‌ها را می‌توان با شباهت وزن داد).
- **انتشار برچسب (label propagation):** همه‌ی تصاویر، چه برچسب‌دار و چه بدون برچسب، رأس‌های یک گراف‌اند و هر تصویر به k همسایه‌ی شبیه‌ترش وصل می‌شود. برچسب‌های معلوم در چند تکرار روی یال‌ها پخش می‌شوند: هر رأس ترکیبی از برچسب همسایه‌هایش را می‌گیرد و رأس‌های برچسب‌دار برچسب خودشان را حفظ می‌کنند. به این ترتیب برچسب از راه زنجیره‌ای از تصاویر مشابه به تصاویر دورتر هم می‌رسد. یک صورت متداول، تکرار رابطه‌ی زیر است ($S$ ماتریس مجاورت نرمال‌شده‌ی گراف، $Y$ برچسب‌های اولیه، $0<\alpha<1$):

</div>

$$
F \leftarrow \alpha\, S\, F + (1-\alpha)\, Y
$$

<div dir="rtl" style="text-align: right; font-family: Tahoma, 'IRANSans', 'Vazirmatn', 'Segoe UI', sans-serif;">

- **آموزش روی مجموعه‌ی برچسب کاذب (pseudo-labeled pool):** پیش‌بینی‌های مطمئن‌تر روی تصاویر بدون برچسب را «برچسب کاذب» بگیرید. با هزاران برچسب کاذب یک مجموعه‌ی آموزشی بزرگ دارید و می‌توانید روی آن هر مدلی را آموزش دهید: یک CNN از صفر، یا تنظیم دقیق (fine-tune) کامل یک مدل ازپیش‌آموزش‌دیده. اینکه کدام پیش‌بینی‌ها را بپذیرید (آستانه‌ی اطمینان، تعادل بین کلاس‌ها) مهم است، چون مدل خطاهای برچسب کاذب را هم یاد می‌گیرد.
- **بازسازی بازنمایی (re-embed):** مدلی که روی برچسب‌های کاذب تنظیم دقیق شده، دیگر یک مدل ImageNet نیست و ویژگی‌های این دامنه را یاد گرفته است. با آن دوباره برای همه‌ی تصاویر بردار بسازید.
- **انتشار دوباره‌ی برچسب (re-propagate):** روی بردارهای جدید گراف را دوباره بسازید و برچسب‌های **واقعی** را دوباره منتشر کنید. خروجی می‌تواند با پیش‌بینی‌های قبلی ترکیب شود یا برچسب‌های کاذب بهتری برای یک دور دیگر آموزش بسازد.

## کد پایه

کد زیر یک CNN کوچک را **از صفر، بدون افزایش داده**، فقط روی ۲۲۵ تصویر برچسب‌دار آموزش می‌دهد و برای همه‌ی تصاویر آزمون پیش‌بینی می‌سازد. همین حالا کار می‌کند و قابل ارسال است. آن را جایگزین یا بهتر کنید؛ کافی است در پایان، متغیر `predictions` برای همه‌ی تصاویر آزمون (به همان ترتیب `test_ids`) یک شماره‌ی کلاس بین ۰ و ۴۴ داشته باشد.

</div>

```python
# ---- Baseline: small CNN trained from scratch on the 225 labeled images only, no augmentation ----
class SmallCNN(nn.Module):
    def __init__(self, n_classes):
        super().__init__()
        def block(cin, cout):
            return nn.Sequential(
                nn.Conv2d(cin, cout, 3, padding=1, bias=False), nn.BatchNorm2d(cout), nn.ReLU(),
                nn.Conv2d(cout, cout, 3, padding=1, bias=False), nn.BatchNorm2d(cout), nn.ReLU(),
                nn.MaxPool2d(2))
        self.f = nn.Sequential(block(3, 32), block(32, 64), block(64, 128), block(128, 256),
                               nn.AdaptiveAvgPool2d(1), nn.Flatten())
        self.h = nn.Sequential(nn.Dropout(0.3), nn.Linear(256, n_classes))

    def forward(self, x):
        return self.h(self.f(x))


torch.manual_seed(0); np.random.seed(0)
AMP = USE_GPU
t0 = time.time()
model = SmallCNN(N_CLASSES).to(DEVICE)
STEPS, BS, LR = 500, 32, 3e-3
opt = torch.optim.AdamW(model.parameters(), lr=LR, weight_decay=5e-4)
sched = torch.optim.lr_scheduler.OneCycleLR(opt, max_lr=LR, total_steps=STEPS, pct_start=0.15)
scaler = torch.amp.GradScaler("cuda", enabled=AMP)

x_lab = to_tensor(X_lab, size=64)                      # images are downsized to 64x64 for this small CNN
y_lab_t = torch.as_tensor(y_lab, device=DEVICE)
model.train(); step = 0
while step < STEPS:
    perm = torch.randperm(len(x_lab), device=DEVICE)
    for s in range(0, len(perm), BS):
        b = perm[s:s + BS]
        if len(b) < 2:
            continue
        with torch.autocast("cuda", dtype=torch.float16, enabled=AMP):
            loss = F.cross_entropy(model(x_lab[b]).float(), y_lab_t[b])
        opt.zero_grad(set_to_none=True); scaler.scale(loss).backward(); scaler.step(opt); scaler.update()
        sched.step(); step += 1
        if step >= STEPS:
            break

# Predict every test image (same order as test_ids)
model.eval(); preds = []
with torch.inference_mode(), torch.autocast("cuda", dtype=torch.float16, enabled=AMP):
    for s in range(0, len(X_test), 512):
        preds.append(model(to_tensor(X_test[s:s + 512], size=64)).argmax(1).cpu())
predictions = torch.cat(preds).numpy()
print(f"baseline done in {time.time() - t0:.0f}s; distinct predicted classes: {len(np.unique(predictions))}")
```
