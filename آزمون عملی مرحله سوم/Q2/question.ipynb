## 1️⃣ شرح مسئله واقعی

هر روز میلیون‌ها تراکنش آنلاین در سامانه‌های بانکی ثبت می‌شود. واحد مدیریت ریسک یک بانک باید در همان لحظه‌ی انجام تراکنش تشخیص دهد که آیا این تراکنش توسط صاحب واقعی حساب انجام شده (**Class 0: Legitimate**) یا نشانه‌ای از یک **تلاش کلاهبردارانه (Class 1: Fraud)** است.

سناریوی رفتاری: کلاهبرداری آنلاین لزوماً یک الگوی واحد ندارد. در برخی نشست‌ها ممکن است مبلغ تراکنش بزرگ‌تر، تلاش‌های ناموفق ورود بیشتر، یا جابه‌جایی جغرافیایی سریع‌تر دیده شود. در برخی دیگر ممکن است نشانه‌هایی مانند رفتار شبیه «آزمایش کارت»، گیرنده‌های جدید، یا فعالیت در ساعات نامتعارف ظاهر شود.

این رفتارها الزاماً مستقل از هم نیستند؛ ممکن است از چند الگوی رفتاری متفاوت ناشی شوند و در ویژگی‌های مختلف با شدت‌های متفاوت بروز کنند.

داده‌های در دسترس شامل چندین ویژگی رفتاری و تراکنشی برای هر نشست کاربر است، با واحدها و مقیاس‌های کاملاً متفاوت، و متغیر هدف به‌صورت باینری برچسب‌گذاری شده است.

پرسش اصلی این است:

> چگونه می‌توان یک مدل مبتنی بر فاصله را طوری طراحی کرد که با وجود تعداد بالای ویژگی‌ها، هنوز بتواند مرز درستی بین تراکنش سالم و کلاهبردارانه رسم کند؟

دو زاویه نگاه اصلی برای مواجهه با این مسئله وجود دارد:

- نگاه **محلی / مبتنی بر حافظه در فضای ویژگی‌ها**: بررسی شباهت هر تراکنش جدید با نمونه‌های گذشته (سالم و کلاهبردارانه) و تصمیم‌گیری بر اساس نزدیک‌ترین همسایگی.
- نگاه **کاهش بعد**: فشرده‌سازی فضای ویژگی‌ها به ویژگی‌های مؤثر.

> در بخش‌های بعدی، برای ثبت کف مهارت فنی، پیاده‌سازی دو ابزار پایه (**KNN** و **PCA**) از صفر الزامی است؛ اما پس از آن، فضای توسعه باز است.

## 3️⃣ شیوه‌ی نمره‌دهی و معیارهای ارزیابی

ارزیابی شما **تنها بر اساس معیار Macro-F1** روی داده‌های آزمون مخفی انجام می‌شود.

در این معیار، ابتدا برای هر دو کلاس به‌صورت جداگانه `F1-Score` محاسبه می‌شود و سپس میانگین آن‌ها به عنوان معیار نهایی در نظر گرفته می‌شود:

$$
\text{Macro-F1} = \frac{F1_{\text{Legitimate}} + F1_{\text{Fraud}}}{2}
$$

### 📊 فرمول محاسبه نمره نهایی (سقف ۲۰ نمره)

نمره‌ی شما بر اساس عملکرد پیش‌بینی‌های ارسال‌شده در فایل `submission.csv` به‌صورت زیر محاسبه می‌شود. توجه داشته باشید که **کسب نمره‌ی مثبت منوط به عبور از مدل پایه است**:

> ⚠️ **نکته:** در ادامه سؤال، شما ابتدا با یک ایده‌ی اولیه نتیجه را می‌سنجید و سپس روش دیگری را توسعه می‌دهید. درصورتی‌که عملکرد روش دوم شما پایین‌تر از روش اصلی شد، نام تابع مدل اولیه را در بخش انتهایی بنویسید. معیار نمره‌دهی، عملکرد مدلی است که در بخش آخر معرفی شده است.

<div dir="rtl" style="direction:ltr;text-align:right;unicode-bidi:embed">

$$
\text{Score} =
\begin{cases}
0 & \text{if } \text{Macro-F1}_{\text{test}} < \text{Macro-F1}_{\text{baseline}} \\
5 + 15 \times
\left(
\frac{\text{Macro-F1}_{\text{test}} - \text{Macro-F1}_{\text{baseline}}}
{\text{Macro-F1}_{\text{target}} - \text{Macro-F1}_{\text{baseline}}}
\right)
& \text{if } \text{Macro-F1}_{\text{baseline}} \le \text{Macro-F1}_{\text{test}} < \text{Macro-F1}_{\text{target}} \\
20 & \text{if } \text{Macro-F1}_{\text{test}} \ge \text{Macro-F1}_{\text{target}}
\end{cases}
$$

</div>

- **نمره ۰:** کسب Macro-F1 کمتر از مدل پایه (عدم موفقیت در عبور از حداقل‌ها)
- **نمره ۵ (Baseline):** کسب Macro-F1 برابر با مدل پایه‌ی ساده
  $$
  \text{Macro-F1} \approx 0.45
  $$

- **نمره ۲۰ (Target):** رسیدن به Macro-F1 هدف با خط‌لوله‌ی بهینه
  $$
  \text{Macro-F1} \ge 0.90
  $$
</div>
