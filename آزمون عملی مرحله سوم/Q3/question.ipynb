## 1️⃣ شرح مسئله واقعی

در یک کارخانهٔ فولاد، ورق‌های نورد گرم روی خط تولید حرکت می‌کنند و یک دوربین صنعتی از سطح
آن‌ها تصویر خاکستری می‌گیرد. واحد کنترل کیفیت باید بر اساس این تصویر، نوع عیب سطحی را
تشخیص دهد تا ورق‌های معیوب پیش از ارسال به مشتری از خط تولید جدا شوند.

شش نوع عیب ممکن است دیده شود: `crazing` (ترک‌مویی)، `inclusion` (ادخال)، `patches` (لکه)،
`pitted_surface` (حفره‌دار)، `rolled-in_scale` (پوسته‌فرورفته) و `scratches` (خراش). دادهٔ این
سؤال از دیتاست واقعی **NEU Surface Defect Database** گرفته شده است.

مأموریت شما: یک مدل پرسپترون چند لایه (به همراه تکنیک‌های مختلف) بسازید که تصویر سطح را بگیرد و نوع عیب را برگرداند.

## 4️⃣ شیوه نمره‌دهی و معیارهای ارزیابی

ارزیابی شما **تنها بر اساس معیار Macro F1-Score** (میانگین F1 هر ۶ کلاس با وزن یکسان) روی
برچسب واقعی تصاویر آزمون -- که نزد شما نیست -- انجام می‌شود.

### 📊 فرمول محاسبه نمره نهایی (سقف ۲۰ نمره)

نمرهٔ شما به‌صورت **خطی** بر اساس نسبت $\text{F1}$ شما به $\text{F1}$ هدف محاسبه می‌شود.
در این روش، هر بهبود در $\text{F1}$ مستقیماً به نمرهٔ بیشتر تبدیل می‌شود:

<div dir="rtl" style="direction:ltr;text-align:right;unicode-bidi:embed">

$$\text{Score} = \min\left(20,\; 20 \times \frac{\text{F1}_{\text{test}}}{\text{F1}_{\text{target}}}\right)$$

</div>

> **توجه:** این نمرات بر اساس $\text{F1}$ روی **برچسب واقعیِ پنهانِ تصاویر آزمون** محاسبه می‌شوند،
> نه روی داده‌های اعتبارسنجی داخلی شما.


> **توجه:** **$\text{F1}_{\text{target}}$ (هدف):** $0.897$ — رسیدن به این مقدار یا بالاتر، نمرهٔ کامل ۲۰ را می‌دهد.

### محدودیت‌های مساله

> ⚠️ **توجه: در صورت عدم رعایت محدودیت‌های سوال نمره‌ی شما صفر خواهد شد و از فرمول بالا محاسبه نخواهد شد. به عنوان مثال، محدودیت کتابخانه‌ها و مدل‌های و متودهای مجاز، به دست آوردن نتایج از راهی به جز آموزش مدل ویا عدم تطابق فایل csv بارگذاری شده با فایل خروجی نوت‌بوک سوال.**

## 7️⃣ مدل MLP
در کلاس MLP شما باید مدل پرسپترون خود و مسیر شبکه را بنویسید. تابع `train_mlp` یک تابع آماده برای آموزش مدل شماست و پیچیدگی‌های آموزش را برای شما به صورت آماده قرار دادیم. می‌توانید با ورودی دادن مدل خود به این تابع و تعیین پارامترهای ورودی‌اش، مدل خود را آموزش دهید. تابع `mlp_predict` نیز تابعی آماده برای پیش‌بینی مدل شماست.

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import TensorDataset, DataLoader


# ============================================================
# Model — طراحی معماری بر عهده‌ی شماست
# ============================================================

class MLP(nn.Module):
    """
    معماری شبکه‌ی خود را اینجا طراحی کنید.
    
    ورودی: بردار ویژگی با ابعاد سایز ورودی
    خروجی: بردار با ابعاد تعداد کلاس‌ها

    می‌توانید از هر ترکیبی از لایه‌های زیر استفاده کنید:
    - nn.Linear(in_features, out_features)
    - nn.ReLU() / nn.LeakyReLU() / nn.Tanh() / nn.Sigmoid()
    - nn.Dropout(p)
    - nn.BatchNorm1d(num_features)
    """

    def __init__(self, input_size):
        super().__init__()

        # TODO: لایه‌های شبکه را اینجا تعریف کنید
        # مثال:
        # self.net = nn.Sequential(
        #     nn.Linear(input_size, 256),
        #     nn.ReLU(),
        #     nn.Dropout(0.3),... and continue
        # )
        pass

    def forward(self, x):
        # TODO: مسیر عبور داده از شبکه را بنویسید
        # مثال:
        # return self.net(x)
        pass


# ============================================================
# Training — این بخش آماده است، تغییر ندهید
# ============================================================

def train_mlp(model, X_train, y_train, X_val=None, y_val=None,
              weight_decay=0.0, epochs=30, batch_size=64,
              lr=1e-3, seed=0, lr_schedule=False, patience=15):
    """
    آموزش مدل روی داده‌ی آموزشی (با توقف زودهنگام در صورت ارائه X_val/y_val)
    """
    torch.manual_seed(seed)
    
    device = torch.device("cpu")  # همیشه CPU برای سازگاری با سرور داوری
    
    model = model.to(device)
    
    opt = torch.optim.Adam(model.parameters(), lr=lr, weight_decay=weight_decay)
    scheduler = None
    if lr_schedule:
        scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
            opt, mode="min", factor=0.5, patience=3
        )
    
    ds = TensorDataset(
        torch.tensor(X_train, dtype=torch.float32),
        torch.tensor(y_train, dtype=torch.long),
    )
    loader = DataLoader(ds, batch_size=batch_size, shuffle=True, pin_memory=False)
    
    # Early Stopping
    best_state = None
    best_f1 = -1.0
    patience_counter = 0
    use_es = (X_val is not None) and (y_val is not None)
    
    if use_es:
        X_val_t = torch.tensor(X_val, dtype=torch.float32).to(device)
        y_val_np = np.asarray(y_val)
    
    model.train()
    for epoch in range(epochs):
        epoch_loss = 0.0
        for xb, yb in loader:
            xb, yb = xb.to(device), yb.to(device)
            opt.zero_grad()
            loss = F.cross_entropy(model(xb), yb)
            loss.backward()
            opt.step()
            epoch_loss += loss.item() * len(xb)
        
        if scheduler is not None:
            scheduler.step(epoch_loss / len(ds))
        
        # Early Stopping check
        if use_es:
            model.eval()
            with torch.no_grad():
                val_logits = model(X_val_t)
                val_preds = val_logits.argmax(1).cpu().numpy()
            from sklearn.metrics import f1_score as _f1
            val_f1 = _f1(y_val_np, val_preds, average='macro', zero_division=0)
            
            if val_f1 > best_f1 + 1e-4:
                best_f1 = val_f1
                best_state = {k: v.cpu().clone() for k, v in model.state_dict().items()}
                patience_counter = 0
            else:
                patience_counter += 1
                if patience_counter >= patience:
                    break
            model.train()
    
    if use_es and best_state is not None:
        model.load_state_dict(best_state)
        model.to(device)
    
    model.eval()
    return model


def mlp_predict(model, X):
    """پیش‌بینی کلاس با مدل آموزش‌دیده"""
    device = torch.device("cpu")
    model = model.to(device)
    model.eval()
    
    X_t = torch.tensor(X, dtype=torch.float32).to(device)
    
    with torch.no_grad():
        return model(X_t).argmax(1).cpu().numpy()
```

```python
import time, inspect
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.metrics import f1_score

SC_F1 = 0.897
FORBIDDEN_TOKENS = ["Conv1d", "Conv2d", "Conv3d", "torchvision.models", "transformers",
                    "timm.", "pretrained=True", "resnet", "efficientnet", "vit_"]


def score_from_f1(f1, target=SC_F1):
    """
    محاسبه نمره به‌صورت خطی:
    Score = min(20, 20 × (F1 / target))
    """
    return min(20.0, 20.0 * (f1 / target))


def check_mlp_only(cls):
    """
    بررسی قید فقط بودن شبکه با چند روش جایگزین.
    """
    # روش 1: تلاش برای گرفتن سورس با inspect
    try:
        src = inspect.getsource(cls)
        return [tok for tok in FORBIDDEN_TOKENS if tok in src]
    except (OSError, TypeError):
        pass
    
    # روش 2: در Jupyter، فقط سلول‌هایی که Solution/MLP را تعریف می‌کنند
    try:
        from IPython import get_ipython
        ip = get_ipython()
        if ip is not None:
            history = list(ip.history_manager.get_range())
            
            # فقط سلول‌هایی که کلاس Solution یا MLP را تعریف می‌کنند
            # و سلول‌هایی که FORBIDDEN_TOKENS را تعریف می‌کنند را نادیده بگیر
            relevant_code_parts = []
            for _, _, cell_code in history:
                # نادیده گرفتن سلول‌های شبیه‌ساز داوری
                if 'FORBIDDEN_TOKENS' in cell_code:
                    continue
                # فقط سلول‌هایی که راه‌حل را تعریف می‌کنند
                if 'class Solution' in cell_code or 'class MLP' in cell_code:
                    relevant_code_parts.append(cell_code)
            
            if relevant_code_parts:
                relevant_code = '\n'.join(relevant_code_parts)
                found = [tok for tok in FORBIDDEN_TOKENS if tok in relevant_code]
                return found
    except Exception:
        pass
    
    # روش 3: بررسی ساختار لایه‌های مدل (fallback)
    try:
        import torch.nn as nn
        
        forbidden_layers = []
        import sys
        main_ns = sys.modules.get('__main__')
        if main_ns and hasattr(main_ns, 'MLP'):
            MLP_cls = getattr(main_ns, 'MLP')
            try:
                sample_model = MLP_cls(100)
                for name, module in sample_model.named_modules():
                    if isinstance(module, (nn.Conv1d, nn.Conv2d, nn.Conv3d)):
                        forbidden_layers.append(name)
            except Exception:
                pass
        
        return forbidden_layers if forbidden_layers else []
    except Exception:
        pass
    
    return None


# ─────────────────────────────────────────────────────
# بررسی قید شبکه
# ─────────────────────────────────────────────────────

hits = check_mlp_only(Solution)
if hits:
    print(f"قید «فقط شبکه»: مردود — یافت شد: {hits}")
elif hits is not None:
    print("قید «فقط شبکه»: قبول ✓")

# ─────────────────────────────────────────────────────
# تقسیم داده به آموزش و اعتبارسنجی
# ─────────────────────────────────────────────────────
tr_idx, va_idx = train_test_split(
    np.arange(len(images)),
    test_size=0.2,
    random_state=1,
    stratify=labels
)

# ─────────────────────────────────────────────────────
# اجرای fit و predict
# ─────────────────────────────────────────────────────

t0 = time.time()
sol = Solution()
sol.fit({"images": images[tr_idx], "labels": labels[tr_idx]})
pred = np.asarray(sol.predict({"images": images[va_idx]}))
elapsed = time.time() - t0

# ─────────────────────────────────────────────────────
# اعتبارسنجی خروجی
# ─────────────────────────────────────────────────────

assert len(pred) == len(va_idx), "طول خروجی نادرست"
assert set(np.unique(pred)) <= set(range(NUM_CLASSES)), "برچسب نامعتبر"

# ─────────────────────────────────────────────────────
# محاسبه معیار
# ─────────────────────────────────────────────────────

f1 = f1_score(labels[va_idx], pred, average="macro", zero_division=0)
print(f"Macro F1 تخمینی        : {f1:.4f}")
print(f"نمرهٔ تخمینی            : {score_from_f1(f1):.1f} / 20")
print(f"زمان اجرا (fit+predict) : {elapsed:.1f} ثانیه  (سقف: 600 ثانیه)")
```
