<p style="direction:rtl;text-align:right;margin:0 0 0.85em;line-height:1.95;">قبل از هر تحلیلی باید مطمئن شویم هیچ مسافری دوبار در دیتاست ثبت نشده — یک سطر تکراری می‌تواند میانگین‌ها و نرخ‌ها (مثل نرخ بقا) را به‌اشتباه دوبار وزن بدهد.</p>

```python
n_dup = df.duplicated().sum()
print("exact duplicate rows:", n_dup)

df = df.drop_duplicates().reset_index(drop=True)
print("shape after dropping duplicates:", df.shape)
```

<div dir="rtl" style="direction:rtl;text-align:right;line-height:1.95;font-size:15px;font-family:Tahoma,'IRANSans','Vazirmatn','Segoe UI',sans-serif;">
<p style="direction:rtl;text-align:right;margin:0 0 0.85em;line-height:1.95;"><strong>نتیجه‌گیری.</strong> هیچ سطر کاملاً تکراری در این دیتاست پیدا نشد (<code style="direction:ltr;text-align:left;unicode-bidi:embed;display:inline-block;background:rgba(128,128,128,0.18);padding:0 4px;border-radius:3px;font-family:monospace;">۰</code> مورد). بخشی از دلیلش این است که دیتاست ما ستون شناسهٔ یکتا (id) ندارد، ولی ترکیب اسم، سن، بلیط و کرایه تقریباً همیشه هر مسافر را از بقیه متمایز می‌کند. با این حال این چک را هرگز نباید حذف کنی — در دیتاست‌های واقعی (خصوصاً وقتی از چند منبع جمع‌آوری یا چند بار وارد شده‌اند) پیدا کردن و حذف سطرهای تکراری معمولاً اولین قدم قبل از هر تحلیل دیگری است.</p>
</div>

```python
missing_pct = (df.isna().mean() * 100).round(1)

missing_pct.sort_values(ascending=False).plot(kind="barh", color="#4C72B0")
plt.xlabel("Missing Data Percentage")
plt.title("Missing Data Map in Titanic Dataset")
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()

missing_pct.sort_values(ascending=False)
```

<ul style="direction:rtl;text-align:right;margin:0 0 0.85em;padding-right:1.6em;padding-left:0;list-style-position:outside;">
<li style="direction:rtl;text-align:right;margin:0.25em 0;line-height:1.9;"><strong>z-score:</strong> فاصلهٔ هر مقدار از میانگین، بر حسب انحراف معیار — <code style="direction:ltr;text-align:left;unicode-bidi:embed;display:inline-block;background:rgba(128,128,128,0.18);padding:0 4px;border-radius:3px;font-family:monospace;">z = (x − mean) / std</code>؛ معمولاً <code style="direction:ltr;text-align:left;unicode-bidi:embed;display:inline-block;background:rgba(128,128,128,0.18);padding:0 4px;border-radius:3px;font-family:monospace;">|z| &gt; 3</code> را پرت در نظر می‌گیرند.</li>
<li style="direction:rtl;text-align:right;margin:0.25em 0;line-height:1.9;"><strong>IQR (دامنهٔ میان‌چارکی):</strong> فاصلهٔ بین چارک سوم (Q3) و چارک اول (Q1)؛ هر مقدار بیرون از <code style="direction:ltr;text-align:left;unicode-bidi:embed;display:inline-block;background:rgba(128,128,128,0.18);padding:0 4px;border-radius:3px;font-family:monospace;">[Q1 − 1.5×IQR, Q3 + 1.5×IQR]</code> را پرت در نظر می‌گیرند.</li>
</ul>
<p style="direction:rtl;text-align:right;margin:0 0 0.85em;line-height:1.95;"><strong>حدس بزن:</strong> فکر می‌کنی این دو روش روی ستون <code style="direction:ltr;text-align:left;unicode-bidi:embed;display:inline-block;background:rgba(128,128,128,0.18);padding:0 4px;border-radius:3px;font-family:monospace;">fare</code> عدد نسبتاً یکسانی از پرت‌ها می‌دهند، یا خیلی با هم فرق دارند؟</p>
</div>

```python
def count_outliers(x):
    z = (x - x.mean()) / x.std()
    n_z = (z.abs() > 3).sum()
    Q1, Q3 = x.quantile(0.25), x.quantile(0.75)
    IQR = Q3 - Q1
    lo, hi = Q1 - 1.5 * IQR, Q3 + 1.5 * IQR
    n_iqr = ((x < lo) | (x > hi)).sum()
    return n_z, n_iqr, (round(float(lo), 2), round(float(hi), 2))

for col in ["fare", "age_filled", "family_size"]:
    n_z, n_iqr, bounds = count_outliers(df[col])
    print(f"{col:12s}  z-score>3: {n_z:4d} ({n_z/len(df)*100:4.1f}%)   IQR: {n_iqr:4d} ({n_iqr/len(df)*100:4.1f}%)   IQR bounds={bounds}")
```

</div>

<div dir="rtl" style="direction:rtl;text-align:right;line-height:1.95;font-size:15px;font-family:Tahoma,'IRANSans','Vazirmatn','Segoe UI',sans-serif;">
<p style="direction:rtl;text-align:right;margin:0 0 0.85em;line-height:1.95;"><strong>نتیجه‌گیری.</strong> روی <code style="direction:ltr;text-align:left;unicode-bidi:embed;display:inline-block;background:rgba(128,128,128,0.18);padding:0 4px;border-radius:3px;font-family:monospace;">fare</code> خام، z-score فقط ۳۸ پرت (۲٫۹٪) پیدا می‌کند ولی IQR <strong>۱۷۱ پرت (۱۳٫۱٪)</strong> — یعنی این دو روش اصلاً با هم توافق ندارند! دلیلش این است که <code style="direction:ltr;text-align:left;unicode-bidi:embed;display:inline-block;background:rgba(128,128,128,0.18);padding:0 4px;border-radius:3px;font-family:monospace;">fare</code> به‌شدت کج است: انحراف معیار (که z-score به آن وابسته است) توسط چند مقدار خیلی بزرگ متورم می‌شود و z-score را نسبت به مقادیر بزرگ «بی‌حس‌تر» می‌کند، در حالی‌که IQR فقط به چارک‌ها وابسته است و حساس‌تر می‌ماند.</p>
<p style="direction:rtl;text-align:right;margin:0 0 0.85em;line-height:1.95;">نکتهٔ مهم‌تر: وقتی نگاه می‌کنیم این ۱۷۱ «پرت» IQR روی <code style="direction:ltr;text-align:left;unicode-bidi:embed;display:inline-block;background:rgba(128,128,128,0.18);padding:0 4px;border-radius:3px;font-family:monospace;">fare</code> خام چه کسانی هستند، <strong>۱۵۳ نفر (۸۹٪) درجه‌یک</strong> هستند — یعنی این «پرت‌ها» عمدتاً بلیط‌های گران‌قیمت واقعی‌اند، نه خطای داده! اگر همهٔ آن‌ها را کورکورانه حذف کنیم، عملاً بیشتر مسافران درجه‌یک را از دیتاست پاک کرده‌ایم. با <code style="direction:ltr;text-align:left;unicode-bidi:embed;display:inline-block;background:rgba(128,128,128,0.18);padding:0 4px;border-radius:3px;font-family:monospace;">log_fare</code> تعداد پرت‌ها به‌شدت کم می‌شود (z-score: ۲۱، IQR: ۴۲) — دوباره همان درسِ بخش قبل: تبدیل لگاریتمی، «پرت بودنِ کاذب» ناشی از چولگی را برطرف می‌کند.</p>
<p style="direction:rtl;text-align:right;margin:0 0 0.85em;line-height:1.95;">روی <code style="direction:ltr;text-align:left;unicode-bidi:embed;display:inline-block;background:rgba(128,128,128,0.18);padding:0 4px;border-radius:3px;font-family:monospace;">family_size</code> هم IQR (~۹٫۵٪) خیلی بیشتر از z-score (~۲٫۷٪) پرت نشان می‌دهد — چون <code style="direction:ltr;text-align:left;unicode-bidi:embed;display:inline-block;background:rgba(128,128,128,0.18);padding:0 4px;border-radius:3px;font-family:monospace;">family_size</code> یک متغیر گسسته و کج است (بیشتر مقادیر ۱ یا ۲ هستند، پس IQR خیلی کوچک می‌شود و هر خانوادهٔ ۴ نفره یا بیشتر را «پرت» می‌شمارد)، در حالی‌که داشتن خانوادهٔ ۴ یا ۵ نفره اصلاً چیز عجیبی نیست.</p>
<p style="direction:rtl;text-align:right;margin:0 0 0.85em;line-height:1.95;"><strong>قاعدهٔ کلی: پرت آماری بودن، به‌معنی خطا بودن نیست.</strong> قبل از حذف هر پرتی، باید فهمید <em>چرا</em> آن مقدار پرت است — گاهی یک زیرگروه واقعی است (مثل درجه‌یک‌ها)، گاهی نتیجهٔ چولگی توزیع است (که با تبدیل لگاریتمی حل می‌شود)، و گاهی واقعاً یک خطای ثبت داده است که باید حذف یا اصلاح شود.</p>

<h2 style="direction:rtl;text-align:right;font-size:1.35em;font-weight:700;margin:1.1em 0 0.4em;line-height:1.6;">بخش ۱۹ — یک ستون دیگر: ساعت سوار شدن</h2>
<p style="direction:rtl;text-align:right;margin:0 0 0.85em;line-height:1.95;">هنوز به سراغ ستون <code style="direction:ltr;text-align:left;unicode-bidi:embed;display:inline-block;background:rgba(128,128,128,0.18);padding:0 4px;border-radius:3px;font-family:monospace;">boarding_hour</code> نرفته‌ایم — ساعتی که هر مسافر سوار کشتی شده (بر مبنای ۲۴ ساعت؛ مثلاً ۲۳٫۵ یعنی ساعت ۱۱ و نیم شب).</p>
<p style="direction:rtl;text-align:right;margin:0 0 0.85em;line-height:1.95;"><strong>حدس بزن:</strong> فکر می‌کنی ساعت سوار شدن با طبقهٔ بلیط (<code style="direction:ltr;text-align:left;unicode-bidi:embed;display:inline-block;background:rgba(128,128,128,0.18);padding:0 4px;border-radius:3px;font-family:monospace;">pclass</code>) رابطه دارد؟</p>
</div>

```python
arith_mean = df.groupby("pclass")["boarding_hour"].mean().round(2)
arith_mean
```

<div dir="rtl" style="direction:rtl;text-align:right;line-height:1.95;font-size:15px;font-family:Tahoma,'IRANSans','Vazirmatn','Segoe UI',sans-serif;">
<p style="direction:rtl;text-align:right;margin:0 0 0.85em;line-height:1.95;"><strong>نتیجه‌گیری اول.</strong> میانگین: درجه‌یک <strong>۱۰٫۱۵</strong>، درجه‌دو <strong>۱۴٫۸۴</strong>، درجه‌سه <strong>۱۴٫۷۳</strong>. درجه‌دو و درجه‌سه تقریباً <strong>یک عدد یکسان</strong> دارند — بر اساس این میانگین، انگار ساعت سوار شدن هیچ فرق معناداری بین این دو گروه نمی‌گذارد.</p>
<p style="direction:rtl;text-align:right;margin:0 0 0.85em;line-height:1.95;">بیا این را واقعاً امتحان کنیم: اگر بخواهیم <em>فقط</em> از روی <code style="direction:ltr;text-align:left;unicode-bidi:embed;display:inline-block;background:rgba(128,128,128,0.18);padding:0 4px;border-radius:3px;font-family:monospace;">boarding_hour</code> حدس بزنیم مسافر در کدام کلاس بوده — با ساده‌ترین روش ممکن (فاصلهٔ ساعت هر مسافر تا میانگین هر کلاس، و انتخاب نزدیک‌ترین) — این ویژگی چقدر خوب کار می‌کند؟</p>
</div>

```python
def naive_predict(h):
    dists = {pc: abs(h - c) for pc, c in arith_mean.items()}
    return min(dists, key=dists.get)

df["pred_naive"] = df["boarding_hour"].apply(naive_predict)
acc_naive = (df["pred_naive"] == df["pclass"]).mean()
baseline_acc = (df["pclass"] == df["pclass"].mode()[0]).mean()

print("دقت پیش‌بینی با فاصلهٔ خام روی boarding_hour:", round(acc_naive, 3))
print("دقت یک مدل تنبل (همیشه پرتکرارترین کلاس)    :", round(baseline_acc, 3))
print()
print("جدول درهم‌ریختگی (سطر=واقعی، ستون=پیش‌بینی):")
print(pd.crosstab(df["pclass"], df["pred_naive"]))
```

<div dir="rtl" style="direction:rtl;text-align:right;line-height:1.95;font-size:15px;font-family:Tahoma,'IRANSans','Vazirmatn','Segoe UI',sans-serif;">
<p style="direction:rtl;text-align:right;margin:0 0 0.85em;line-height:1.95;"><strong>نتیجه‌گیری — این ویژگی نه‌تنها کمک نکرد، بدتر هم کرد.</strong> دقت با فاصلهٔ خام فقط <strong>۳۱٫۸٪</strong> است — حتی از یک مدل تنبل که همیشه پرتکرارترین کلاس را حدس می‌زند (<strong>۵۴٫۲٪</strong>) هم بدتر! در جدول درهم‌ریختگی می‌بینی که تقریباً <strong>هیچ</strong> مسافر درجه‌سه‌ای درست پیش‌بینی نشده (فقط ۲ نفر از ۷۰۹ نفر) — اکثرشان به‌اشتباه به درجه‌دو نسبت داده شده‌اند.</p>
<p style="direction:rtl;text-align:right;margin:0 0 0.85em;line-height:1.95;">چرا؟ چون میانگین‌های خام درجه‌دو (۱۴٫۸۴) و درجه‌سه (۱۴٫۷۳) تقریباً روی هم افتادند، در نتیجه فاصلهٔ خام هیچ‌وقت نمی‌تواند این دو گروه را از هم جدا کند — با این‌که در واقعیت، الگوی زمانی این دو گروه کاملاً متفاوت است.</p>
</div>

<div dir="rtl" style="direction:rtl;text-align:right;line-height:1.95;font-size:15px;font-family:Tahoma,'IRANSans','Vazirmatn','Segoe UI',sans-serif;">
<h3 style="direction:rtl;text-align:right;font-size:1.15em;font-weight:700;margin:1em 0 0.35em;line-height:1.6;">مشکل کجاست؟</h3>
<p style="direction:rtl;text-align:right;margin:0 0 0.85em;line-height:1.95;">ساعت یک متغیر <strong>چرخه‌ای</strong> است: ساعت ۲۳ و ساعت ۰ فقط یک ساعت با هم فاصله دارند، نه ۲۳ ساعت. وقتی مسافرانی که دور نیمه‌شب سوار شده‌اند، بعضی‌شان کمی قبل از ۲۴ افتاده‌اند (مثلاً ۲۳٫۵) و بعضی کمی بعد از ۰ (مثلاً ۰٫۵)، میانگین‌گیری خام آن‌ها را در وسط بازهٔ عددی (بعدازظهر!) می‌گذارد — دقیقاً همان‌جایی که یک گروه کاملاً متفاوت (سوار شدن واقعی بعدازظهر) هم زندگی می‌کند. هر دو گروه، از دید یک مدل مبتنی بر فاصلهٔ خام، غیرقابل‌تشخیص می‌شوند.</p>
<h3 style="direction:rtl;text-align:right;font-size:1.15em;font-weight:700;margin:1em 0 0.35em;line-height:1.6;">راه‌حل را خودت پیدا کن</h3>
<p style="direction:rtl;text-align:right;margin:0 0 0.85em;line-height:1.95;">فکر کن: چطور می‌شود هر ساعت را طوری به عدد تبدیل کرد که ساعت ۲۳ و ساعت ۰ واقعاً «نزدیک هم» باشند، نه در دو سر یک بازهٔ عددی؟</p>
<p style="direction:rtl;text-align:right;margin:0 0 0.85em;line-height:1.95;">راهنمایی‌ها:</p>
<ul style="direction:rtl;text-align:right;margin:0 0 0.85em;padding-right:1.6em;padding-left:0;list-style-position:outside;">
<li style="direction:rtl;text-align:right;margin:0.25em 0;line-height:1.9;">یک ساعت روی یک <strong>دایره</strong> قرار دارد، نه روی یک خط راست. موقعیت یک نقطه روی دایره را معمولاً با <strong>دو</strong> عدد توصیف می‌کنند، نه یک عدد.</li>
<li style="direction:rtl;text-align:right;margin:0.25em 0;line-height:1.9;">توابعی که بلدی و دقیقاً همین کار را می‌کنند (گرفتن یک زاویه و برگرداندن مختصات روی دایرهٔ واحد) کدامند؟</li>
<li style="direction:rtl;text-align:right;margin:0.25em 0;line-height:1.9;">اگر ۲۴ ساعت را معادل یک دور کامل (۳۶۰ درجه، یا ۲π رادیان) در نظر بگیری، ساعت <code style="direction:ltr;text-align:left;unicode-bidi:embed;display:inline-block;background:rgba(128,128,128,0.18);padding:0 4px;border-radius:3px;font-family:monospace;">h</code> به چه زاویه‌ای تبدیل می‌شود؟</li>
</ul>
<p style="direction:rtl;text-align:right;margin:0 0 0.85em;line-height:1.95;"><strong>TODO:</strong> با استفاده از این راهنمایی‌ها، دو ستون جدید <code style="direction:ltr;text-align:left;unicode-bidi:embed;display:inline-block;background:rgba(128,128,128,0.18);padding:0 4px;border-radius:3px;font-family:monospace;">hour_sin</code> و <code style="direction:ltr;text-align:left;unicode-bidi:embed;display:inline-block;background:rgba(128,128,128,0.18);padding:0 4px;border-radius:3px;font-family:monospace;">hour_cos</code> بساز (از <code style="direction:ltr;text-align:left;unicode-bidi:embed;display:inline-block;background:rgba(128,128,128,0.18);padding:0 4px;border-radius:3px;font-family:monospace;">numpy</code> کمک بگیر: <code style="direction:ltr;text-align:left;unicode-bidi:embed;display:inline-block;background:rgba(128,128,128,0.18);padding:0 4px;border-radius:3px;font-family:monospace;">np.sin</code>, <code style="direction:ltr;text-align:left;unicode-bidi:embed;display:inline-block;background:rgba(128,128,128,0.18);padding:0 4px;border-radius:3px;font-family:monospace;">np.cos</code>, <code style="direction:ltr;text-align:left;unicode-bidi:embed;display:inline-block;background:rgba(128,128,128,0.18);padding:0 4px;border-radius:3px;font-family:monospace;">np.pi</code>). سپس همان کاری که در بخش قبل با فاصلهٔ خام کردی را این‌بار با این دو ویژگی تکرار کن: برای هر کلاس مرکز (میانگین <code style="direction:ltr;text-align:left;unicode-bidi:embed;display:inline-block;background:rgba(128,128,128,0.18);padding:0 4px;border-radius:3px;font-family:monospace;">hour_sin</code> و <code style="direction:ltr;text-align:left;unicode-bidi:embed;display:inline-block;background:rgba(128,128,128,0.18);padding:0 4px;border-radius:3px;font-family:monospace;">hour_cos</code>) را حساب کن، هر مسافر را به نزدیک‌ترین مرکز نسبت بده، و دقت را دوباره بسنج.</p>
</div>

```python
df["hour_sin"] = np.sin(2 * np.pi * df["boarding_hour"] / 24)
df["hour_cos"] = np.cos(2 * np.pi * df["boarding_hour"] / 24)

cyc_centroids = df.groupby("pclass")[["hour_sin", "hour_cos"]].mean()

def cyc_predict(row):
    dists = {pc: (row["hour_sin"] - c["hour_sin"]) ** 2 + (row["hour_cos"] - c["hour_cos"]) ** 2
              for pc, c in cyc_centroids.iterrows()}
    return min(dists, key=dists.get)

df["pred_cyc"] = df.apply(cyc_predict, axis=1)
acc_cyc = (df["pred_cyc"] == df["pclass"]).mean()

print("Accuracy with cyclical features (hour_sin, hour_cos):", round(acc_cyc, 3))
print()
print(pd.crosstab(df["pclass"], df["pred_cyc"]))

plt.figure(figsize=(6, 6))
for pc, color in zip([1, 2, 3], ["#4C72B0", "#DD8452", "#55A868"]):
    sub = df[df["pclass"] == pc]
    plt.scatter(sub["hour_sin"], sub["hour_cos"], s=10, alpha=0.5, label=f"pclass {pc}", color=color)
plt.xlabel("Hour Sin")
plt.ylabel("Hour Cos")
plt.title("Boarding Hour on Unit Circle")
plt.legend()
plt.axis("equal")
plt.show()
```

<h3 style="direction:rtl;text-align:right;font-size:1.15em;font-weight:700;margin:1em 0 0.35em;line-height:1.6;">کاربرد واقعی: این مسئله دقیقاً کِی در دنیای واقعی پیش می‌آید؟</h3>
<p style="direction:rtl;text-align:right;margin:0 0 0.85em;line-height:1.95;">هر جا بخواهی از <strong>ساعت وقوع یک رویداد</strong> برای پیش‌بینی چیزی استفاده کنی — ساعت ورود مسافر به فرودگاه، ساعت ثبت یک تراکنش بانکی برای تشخیص تقلب، ساعت شروع یک پیام برای دسته‌بندی نوع کاربر — و از یک مدل مبتنی بر فاصله (مثل k-نزدیک‌ترین‌همسایه یا خوشه‌بندی) یا یک مدل خطی ساده استفاده کنی، دقیقاً همین اتفاق می‌افتد:</p>
<ul style="direction:rtl;text-align:right;margin:0 0 0.85em;padding-right:1.6em;padding-left:0;list-style-position:outside;">
<li style="direction:rtl;text-align:right;margin:0.25em 0;line-height:1.9;">اگر ساعت را <strong>خام</strong> بدهی: دو رویداد که در واقعیت فقط چند دقیقه از هم فاصله دارند ولی دو طرف مرز نیمه‌شب افتاده‌اند (مثلاً ۲۳:۵۰ و ۰۰:۱۰)، برای مدل تقریباً <strong>در دو سر مخالف بازه</strong> به نظر می‌رسند. و اگر یک گروه از رویدادها دقیقاً حول نیمه‌شب متمرکز باشند (مثل الگوی درجه‌سه در همین دیتاست)، میانگین یا مرکز آن گروه به‌اشتباه در وسط روز محاسبه می‌شود و با یک گروه کاملاً متفاوت (که واقعاً وسط روز اتفاق افتاده) قاطی می‌شود.</li>
<li style="direction:rtl;text-align:right;margin:0.25em 0;line-height:1.9;">نتیجه: مدل یا اصلاً یاد نمی‌گیرد ساعت مهم است (سیگنال در میانگین‌گیری گم می‌شود)، یا بدتر، یک الگوی غلط یاد می‌گیرد که دقتش از حدس تصادفی هم پایین‌تر است — دقیقاً همان چیزی که الان با چشم خودت دیدی (۳۱٫۸٪ در برابر ۵۴٫۲٪).</li>
<li style="direction:rtl;text-align:right;margin:0.25em 0;line-height:1.9;">اگر ساعت را با <code style="direction:ltr;text-align:left;unicode-bidi:embed;display:inline-block;background:rgba(128,128,128,0.18);padding:0 4px;border-radius:3px;font-family:monospace;">sin</code>/<code style="direction:ltr;text-align:left;unicode-bidi:embed;display:inline-block;background:rgba(128,128,128,0.18);padding:0 4px;border-radius:3px;font-family:monospace;">cos</code> کدگذاری کنی: فاصلهٔ بین دو ساعت همیشه فاصلهٔ واقعیِ زمانی را نشان می‌دهد، پس هر مدل مبتنی بر فاصله یا خطی می‌تواند از این سیگنال درست استفاده کند — دقیقاً چیزی که با جهش دقت از ۳۱٫۸٪ به ۸۷٫۲٪ دیدی.</li>

<p style="direction:rtl;text-align:right;margin:0 0 0.85em;line-height:1.95;">هر متغیر دیگری که روی یک چرخه تعریف شده — روز هفته، ماه سال، جهت قطب‌نما، زاویهٔ چرخش — همین ریسک را دارد. این دقیقاً همان ایده‌ای است که چند هفتهٔ دیگر در بخش NLP دوباره می‌بینی: ترنسفورمرها برای این‌که مدل بفهمد کلمهٔ اول جمله کجاست، از <strong>positional encoding</strong> با ترکیبی از توابع sin/cos استفاده می‌کنند — دقیقاً همین تکنیکی که همین الان روی ساعت تمرین کردی.</p>
