<div dir="rtl" lang="fa" style="font-family: 'Vazirmatn', sans-serif; line-height: 1.8; direction: rtl; text-align: right;">
<h2 style="text-align: right; color: #3498db;">داستان سفر ما: افق‌های کاربردی شبکه‌های عصبی پیچشی (CNN)</h2>
<p style="text-align: right;">شبکه‌های عصبی پیچشی (CNN) ستون فقرات انقلاب بینایی ماشین در یک دهه اخیر بوده‌اند. از ساده‌ترین تسک‌های تصمیم‌گیری سراسری تا پیچیده‌ترین پردازش‌های مکان‌محور و تفکیک پیکسلی، ساختارهای پیچشی نقشی بی‌بدیل ایفا می‌کنند. در این تمرین عملی و جامع، به بررسی ۶ سناریوی کلیدی و راهبردی در بینایی ماشین می‌پردازیم:</p>
<ol style="text-align: right;">
<li style="text-align: right;"><b>۱. یادگیری انتقالی در برابر آموزش از صفر روی MNIST:</b> ارزیابی مزیت شبکه از‌پیش‌آموزش‌دیده روی Fashion-MNIST و مقایسه سرعت همگرایی، دقت و کارآمدی داده با شبکه‌ای که با وزن‌های تصادفی از ابتدا آموزش می‌بیند.</li>
<li style="text-align: right;"><b>۲. یادگیری انتقالی با MobileNetV3-Small و ساختار ترکیبی (Hybrid Backbone):</b> بررسی چرایی عملکرد ضعیف مدل‌های سنگین ImageNet روی تصاویر کوچک و تک‌کاناله (ناهمخوانی ساختار و دامنه) و پیاده‌سازی معماری بهینه با استفاده از لایه‌های اولیه MobileNet به همراه سر کانولوشنی سفارشی.</li>
<li style="text-align: right;"><b>۳. تنظیم دقیق انتخابی با اطلاعات فیشر روی شبکه کامل و لایه‌های ابتدایی (Full & Shallow Fisher):</b> پیاده‌سازی معیارهای سه‌گانه اهمیت وزن‌ها ($|W|$، حساسیت گرادیان و اطلاعات فیشر) و اعمال انجماد هوشمند پارامترها روی هر دو معماری کامل و ساختار ترکیبی لایه‌های ابتدایی.</li>
<li style="text-align: right;"><b>۴. پایداری در شرایط عدم‌تعادل شدید داده‌ها و مقایسه ۴ راهبرد با F1-Score:</b> ساخت توزیع نامتعادل دم‌بلند و مقایسه جامع ۴ روش (Cross-Entropy استاندارد، Focal Loss، بیش‌نمونه‌برداری Oversampling، و کم‌نمونه‌برداری Downsampling) بر اساس بیشینه امتیاز Macro F1-Score.</li>
<li style="text-align: right;"><b>۵. آشکارسازی تک‌شیء با رگرسیون مستقیم مختصات (Single-Object Detection):</b> بارگذاری داده‌های مصنوعی ذخیره‌شده روی دیسک و آموزش ۲۵ دور مدل آشکارساز با تابع زیان ترکیبی Smooth L1 و Cross-Entropy.</li>
<li style="text-align: right;"><b>۶. قطعه‌بندی سلولی در تصویربرداری پزشکی با U-Net و زیان Dice:</b> بارگذاری تصاویر میکروسکوپی سلولی ذخیره‌شده روی دیسک با پس‌زمینه شدیداً غالب (>۹۰٪) و آموزش معماری CellUNet با تابع زیان Soft Dice Loss و تفکیک نمونه‌ها با اتصال مولفه‌ها.</li>
</ol>
<p style="text-align: right; color: var(--jp-ui-font-color2, #6b7280);">تمام آموزش‌های مدل‌ها مجهز به نوار پیشرفت <code dir="ltr">tqdm</code>، گزارش‌های دقیق به ازای هر دور (Epoch Logs) شامل زیان و متریک‌های آموزش و اعتبارسنجی، و انتخاب بهترین مدل بر اساس زیرمجموعه اعتبارسنجی (Validation Checkpoint) هستند.</p>
</div>

```python
# Setup - read-only cell; do not edit
import random
import math
import time
import json
import os
import copy
from typing import Tuple, List, Dict, Optional
from PIL import Image

import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from scipy import ndimage
from tqdm.auto import tqdm

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader, Subset, WeightedRandomSampler
import torchvision
import torchvision.transforms as transforms
import torchvision.models as models
from torchvision.models import MobileNet_V3_Small_Weights

# Ensure reproducibility across runs
def set_seed(seed: int = 42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)

set_seed(42)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Environment ready. Running on device: {device}")
```

<h2 style="text-align: right; color: #3498db;">فاز ۱ — بخش ۱.۱: بارگذاری مجموعه داده MNIST و افراز داده‌ها (آموزش، اعتبارسنجی و آزمون)</h2>
<p style="text-align: right;">پایگاه داده مشهور <b>MNIST</b> شامل ۷۰,۰۰۰ تصویر تک‌کاناله (سیاه‌وسفید) با ابعاد $28 \times 28$ از ارقام دست‌نویس انگلیسی (۰ تا ۹) است. برای ساخت خط لوله‌ای اصولی و استاندارد، داده‌ها را به سه زیرمجموعه متوازن افراز می‌کنیم:</p>
<ul style="text-align: right;">
<li style="text-align: right;"><b>مجموعه آموزش (Train):</b> ۶۰۰ نمونه متوازن (۶۰ نمونه به ازای هر رقم)</li>
<li style="text-align: right;"><b>مجموعه اعتبارسنجی (Validation):</b> ۲۰۰ نمونه متوازن (۲۰ نمونه به ازای هر رقم) جهت رصد پیشرفت و انتخاب بهترین چک‌پوینت وزن‌ها</li>
<li style="text-align: right;"><b>مجموعه آزمون (Test):</b> ۳۰۰ نمونه متوازن (۳۰ نمونه به ازای هر رقم) جهت ارزیابی نهایی و بی‌طرفانه مدل‌ها</li>
</ul>
<p style="text-align: right;">در سلول زیر داده‌ها را بارگذاری کرده و یک گرید $2 \times 5$ از نمونه ارقام به تصویر می‌کشیم.</p>
</div>

```python
# Phase 1.1: Load MNIST and build balanced Train, Validation, and Test splits
transform_mnist = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.1307,), (0.3081,))
])

# Load MNIST datasets from local directory
full_train_mnist = torchvision.datasets.MNIST(root='./data', train=True, download=True, transform=transform_mnist)
full_test_mnist = torchvision.datasets.MNIST(root='./data', train=False, download=True, transform=transform_mnist)

# Build balanced subsets: 60 train, 20 val, 30 test per class
train_targets = full_train_mnist.targets
test_targets = full_test_mnist.targets

train_indices, val_indices, test_indices = [], [], []
for digit in range(10):
    all_train_idx = (train_targets == digit).nonzero(as_tuple=True)[0]
    train_indices.extend(all_train_idx[:60].tolist())
    val_indices.extend(all_train_idx[60:80].tolist())
    
    all_test_idx = (test_targets == digit).nonzero(as_tuple=True)[0]
    test_indices.extend(all_test_idx[:30].tolist())

mnist_train_sub = Subset(full_train_mnist, train_indices)
mnist_val_sub = Subset(full_train_mnist, val_indices)
mnist_test_sub = Subset(full_test_mnist, test_indices)

train_loader = DataLoader(mnist_train_sub, batch_size=32, shuffle=True)
val_loader = DataLoader(mnist_val_sub, batch_size=32, shuffle=False)
test_loader = DataLoader(mnist_test_sub, batch_size=32, shuffle=False)

print(f"MNIST Data Loaded successfully:")
print(f"- Balanced Training set  : {len(mnist_train_sub)} samples (60 per class)")
print(f"- Balanced Validation set: {len(mnist_val_sub)} samples (20 per class)")
print(f"- Balanced Test set      : {len(mnist_test_sub)} samples (30 per class)")

# Display 2x5 grid of sample digits
fig, axes = plt.subplots(2, 5, figsize=(11, 4.5))
for digit in range(10):
    ax = axes[digit // 5, digit % 5]
    img, lbl = mnist_train_sub[digit * 60]
    ax.imshow(img.squeeze(0).numpy(), cmap='gray')
    ax.set_title(f"Class: {lbl}", fontsize=11)
    ax.axis('off')

plt.suptitle("Sample Images from Balanced MNIST Dataset (Digits 0-9)", fontsize=13)
plt.tight_layout()
plt.show()
```

<div dir="rtl" lang="fa" style="font-family: 'Vazirmatn', sans-serif; line-height: 1.8; direction: rtl; text-align: right;">
<h2 style="text-align: right; color: #3498db;">فاز ۱ — بخش ۱.۲: یادگیری انتقالی از Fashion-MNIST در برابر آموزش از صفر (Scratch) روی MNIST</h2>
<p style="text-align: right;">در این بخش یک معماری مدولار CNN متشکل از ۳ بلوک کانولوشنی استاندارد پیاده‌سازی می‌کنیم:</p>
<ol style="text-align: right;">
<li style="text-align: right;"><b>بلوک ۱:</b> نگاشت کانال ورودی $1 \to 32$ با فیلتر $3 \times 3$، نرمال‌سازی دسته‌ای (BatchNorm)، فعال‌ساز ReLU و لایه تجمیع بیشینه (MaxPool2d) $\to (32, 14, 14)$.</li>
<li style="text-align: right;"><b>بلوک ۲:</b> نگاشت کانال‌ها از $32 \to 64$ با فیلتر $3 \times 3$، BatchNorm، ReLU و MaxPool2d $\to (64, 7, 7)$.</li>
<li style="text-align: right;"><b>بلوک ۳:</b> نگاشت کانال‌ها از $64 \to 128$ با فیلتر $3 \times 3$، BatchNorm، ReLU و تجمیع میانگین سراسری (<code dir="ltr">AdaptiveAvgPool2d((1, 1))</code>) $\to (128, 1, 1)$.</li>
<li style="text-align: right;"><b>سر دسته‌بندی:</b> لایه تمام‌متصل نگاشت $128 \to 10$ برای تولید لاجیت‌های ۱۰ کلاس هدف.</li>
</ol>
<p style="text-align: right;">دو آزمایش با تعداد دور یکسان (۱۰ دور) و شرایط کاملاً برابر اجرا می‌کنیم:</p>
<ul style="text-align: right;">
<li style="text-align: right;"><b>مدل ۱ (Pretrained Fashion-MNIST):</b> وزن‌های از‌پیش‌آموزش‌دیده روی Fashion-MNIST (<code dir="ltr">pretrained_fashion_mnist.pth</code>) را بارگذاری کرده و مدل را روی ۶۰۰ داده MNIST تنظیم دقیق (Fine-Tune) می‌کنیم.</li>
<li style="text-align: right;"><b>مدل ۲ (From Scratch):</b> وزن‌های همان معماری را به صورت تصادفی مقداردهی اولیه کرده و از ابتدا روی همان داده‌ها آموزش می‌دهیم.</li>
</ul>
<p style="text-align: right;">در هر دو آزمایش، مدل در پایان هر دور روی مجموعه اعتبارسنجی ارزیابی شده و بهترین چک‌پوینت وزن‌ها بر اساس بالاترین دقت اعتبارسنجی (<code dir="ltr">best_val_acc</code>) ذخیره و بازیابی می‌شود.</p>
</div>

```python
# Phase 1.2: Standard Convolutional Neural Network (Pretrained vs From Scratch)
class ModularCNN(nn.Module):
    # A modular Convolutional Neural Network for 28x28 grayscale image classification
    def __init__(self, num_classes: int = 10):
        super().__init__()
        self.features = nn.Sequential(
            # Block 1: (1, 28, 28) -> (32, 14, 14)
            nn.Conv2d(1, 32, kernel_size=3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(2, 2),
            # Block 2: (32, 14, 14) -> (64, 7, 7)
            nn.Conv2d(32, 64, kernel_size=3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(2, 2),
            # Block 3: (64, 7, 7) -> (128, 1, 1)
            nn.Conv2d(64, 128, kernel_size=3, padding=1),
            nn.BatchNorm2d(128),
            nn.ReLU(inplace=True),
            nn.AdaptiveAvgPool2d((1, 1))
        )
        self.classifier = nn.Linear(128, num_classes)

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        feat = self.features(x)
        out = torch.flatten(feat, 1)
        return self.classifier(out)

def evaluate_accuracy(model: nn.Module, loader: DataLoader) -> float:
    # Computes top-1 accuracy percentage on a given data loader
    model.eval()
    correct, total = 0, 0
    with torch.no_grad():
        for imgs, labels in loader:
            imgs, labels = imgs.to(device), labels.to(device)
            preds = model(imgs).argmax(dim=1)
            correct += (preds == labels).sum().item()
            total += labels.size(0)
    return (correct / max(1, total)) * 100.0

def train_model_with_val_checkpoint(
    model: nn.Module,
    train_loader: DataLoader,
    val_loader: DataLoader,
    optimizer: torch.optim.Optimizer,
    criterion: nn.Module,
    epochs: int = 10,
    model_name: str = "Model"
) -> Dict[str, List[float]]:
    # Unified training loop with tqdm progress bars, per-epoch loss/acc logs, and best validation checkpoint selection
    history = {"train_loss": [], "train_acc": [], "val_loss": [], "val_acc": []}
    best_val_acc = -1.0
    best_weights = copy.deepcopy(model.state_dict())
    
    print(f"\n{'='*25} Training {model_name} ({epochs} Epochs) {'='*25}")
    for epoch in range(1, epochs + 1):
        model.train()
        running_loss, correct, total = 0.0, 0, 0
        
        pbar = tqdm(train_loader, desc=f"[{model_name}] Epoch {epoch:2d}/{epochs}", leave=False)
        for imgs, labels in pbar:
            imgs, labels = imgs.to(device), labels.to(device)
            optimizer.zero_grad()
            outputs = model(imgs)
            loss = criterion(outputs, labels)
            loss.backward()
            optimizer.step()
            
            running_loss += loss.item() * imgs.size(0)
            preds = outputs.argmax(dim=1)
            correct += (preds == labels).sum().item()
            total += labels.size(0)
            pbar.set_postfix(loss=f"{loss.item():.4f}", acc=f"{100.0 * correct / total:.1f}%")
            
        train_loss = running_loss / total
        train_acc = 100.0 * correct / total
        
        # Validation step
        model.eval()
        val_loss, val_correct, val_total = 0.0, 0, 0
        with torch.no_grad():
            for imgs, labels in val_loader:
                imgs, labels = imgs.to(device), labels.to(device)
                outputs = model(imgs)
                loss = criterion(outputs, labels)
                val_loss += loss.item() * imgs.size(0)
                preds = outputs.argmax(dim=1)
                val_correct += (preds == labels).sum().item()
                val_total += labels.size(0)
                
        val_loss /= max(1, val_total)
        val_acc = 100.0 * val_correct / max(1, val_total)
        
        history["train_loss"].append(train_loss)
        history["train_acc"].append(train_acc)
        history["val_loss"].append(val_loss)
        history["val_acc"].append(val_acc)
        
        print(f"Epoch {epoch:2d}/{epochs} | Train Loss: {train_loss:.4f} - Train Acc: {train_acc:5.2f}% | Val Loss: {val_loss:.4f} - Val Acc: {val_acc:5.2f}%")
        
        if val_acc > best_val_acc:
            best_val_acc = val_acc
            best_weights = copy.deepcopy(model.state_dict())

    # Restore highest performing validation checkpoint
    model.load_state_dict(best_weights)
    print(f"--> Restored Best Validation Checkpoint for {model_name} (Val Acc: {best_val_acc:.2f}%)")
    return history

# ---------------------------------------------------------
# Experiment A: Pretrained Fashion-MNIST Fine-Tuning on MNIST
# ---------------------------------------------------------
set_seed(42)
pretrained_cnn = ModularCNN(num_classes=10).to(device)
pretrained_path = "pretrained_fashion_mnist.pth"

if os.path.exists(pretrained_path):
    state_dict = torch.load(pretrained_path, map_location=device)
    pretrained_cnn.load_state_dict(state_dict)
    print(f"Loaded pretrained weights from {pretrained_path}")
else:
    print(f"Notice: {pretrained_path} not found. Running with standard initialization.")

opt_pre = torch.optim.Adam(pretrained_cnn.parameters(), lr=0.002)
crit = nn.CrossEntropyLoss()
history_pre = train_model_with_val_checkpoint(
    pretrained_cnn, train_loader, val_loader, opt_pre, crit, epochs=10, model_name="Pretrained-CNN"
)
acc_pretrained = evaluate_accuracy(pretrained_cnn, test_loader)
print(f"Final Test Accuracy (Pretrained Fashion-MNIST -> MNIST): {acc_pretrained:.2f}%")

# ---------------------------------------------------------
# Experiment B: Random Initialization (Training From Scratch) on MNIST
# ---------------------------------------------------------
set_seed(42)
scratch_cnn = ModularCNN(num_classes=10).to(device)
opt_scratch = torch.optim.Adam(scratch_cnn.parameters(), lr=0.002)
history_scratch = train_model_with_val_checkpoint(
    scratch_cnn, train_loader, val_loader, opt_scratch, crit, epochs=10, model_name="Scratch-CNN"
)
acc_scratch = evaluate_accuracy(scratch_cnn, test_loader)
print(f"Final Test Accuracy (From Scratch MNIST): {acc_scratch:.2f}%")

# ---------------------------------------------------------
# Comparison Visualization: Loss and Accuracy Curves
# ---------------------------------------------------------
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 4.5))

epochs_range = range(1, 11)
ax1.plot(epochs_range, history_pre["train_loss"], 'b-o', label="Pretrained Train Loss")
ax1.plot(epochs_range, history_pre["val_loss"], 'b--', label="Pretrained Val Loss")
ax1.plot(epochs_range, history_scratch["train_loss"], 'r-s', label="Scratch Train Loss")
ax1.plot(epochs_range, history_scratch["val_loss"], 'r--', label="Scratch Val Loss")
ax1.set_title("Training & Validation Loss Comparison", fontsize=11)
ax1.set_xlabel("Epoch")
ax1.set_ylabel("Loss")
ax1.grid(True, linestyle="--", alpha=0.5)
ax1.legend()

ax2.plot(epochs_range, history_pre["val_acc"], 'b-o', label=f"Pretrained Val Acc (Final Test: {acc_pretrained:.1f}%)")
ax2.plot(epochs_range, history_scratch["val_acc"], 'r-s', label=f"Scratch Val Acc (Final Test: {acc_scratch:.1f}%)")
ax2.set_title("Validation Accuracy Progression", fontsize=11)
ax2.set_xlabel("Epoch")
ax2.set_ylabel("Validation Accuracy (%)")
ax2.grid(True, linestyle="--", alpha=0.5)
ax2.legend()

plt.suptitle("Transfer Learning (Fashion-MNIST Pretrained) vs Training From Scratch on MNIST", fontsize=13)
plt.tight_layout()
plt.show()

print("\n" + "="*65)
print(f"{'Model Configuration':<30} | {'Best Val Acc':<14} | {'Test Accuracy':<14}")
print("="*65)
print(f"{'Pretrained (Fashion-MNIST)':<30} | {max(history_pre['val_acc']):.2f}%{'':<8} | {acc_pretrained:.2f}%")
print(f"{'From Scratch (Random Init)':<30} | {max(history_scratch['val_acc']):.2f}%{'':<8} | {acc_scratch:.2f}%")
print("="*65)
```

<h2 style="text-align: right; color: #3498db;">فاز ۱ — بخش ۱.۳: یادگیری انتقالی با MobileNetV3 و تحلیل چرایی شکست مدل کامل (Domain & Structural Mismatch)</h2>
<p style="text-align: right;">مدل <b>MobileNetV3-Small</b> روی بیش از ۱.۲ میلیون تصویر رنگی طبیعی از پایگاه داده ImageNet آموزش دیده است. هنگامی که این مدل را به صورت کامل برای داده‌های MNIST تنظیم دقیق می‌کنیم، عملکرد آن در مقایسه با CNN ساده بسیار پایین‌تر یا همگرایی آن کندتر است. دلایل بنیادین این پدیده عبارتند از:</p>
<ol style="text-align: right;">
<li style="text-align: right;"><b>ناهمخوانی موضوع و دامنه (Subject / Domain Mismatch):</b> ویژگی‌های آموخته‌شده در ImageNet برای تصاویر رنگی با بافت‌های پیچیده طبیعی، سایه‌روشن و اشیاء واقعی بهینه‌سازی شده‌اند؛ در حالی که MNIST صرفاً شامل خطوط و لبه‌های ساده تک‌کاناله سیاه و سفید است.</li>
<li style="text-align: right;"><b>ناهمخوانی ساختاری و عمق بیش‌از‌حد (Structural Mismatch):</b> ساختار کانولوشن‌های تفکیک‌پذیر عمقی (Depthwise Separable) و کاهش ابعاد مکرر MobileNet برای ابعاد بزرگ ($224 \times 224$) طراحی شده است؛ اعمال آن روی ورودی‌های کوچک $28 \times 28$ باعث فروپاشی سریع رزولوشن مکانی در لایه‌های ابتدایی و بیش‌پارامتری شدن مدل روی داده‌های اندک می‌شود.</li>
</ol>
<h3 style="text-align: right; color: var(--jp-content-font-color1, #1e6bb8);">راهکار بهینه: استخراج ویژگی کم‌عمق + سر سفارشی (Shallow MobileNet + Custom Head)</h3>
<p style="text-align: right;">برای بهره‌برداری واقعی از MobileNet، صرفاً <b>لایه‌های ابتدایی</b> آن (که فیلترهای پایه لبه و کنتراست را استخراج می‌کنند) را حفظ کرده و سپس یک ساختار کانولوشنی سفارشی متناسب با تصاویر کوچک به آن متصل می‌کنیم. در سلول زیر، هر دو مدل را با همان تعداد دور یکسان (۱۰ دور) آموزش داده و نتایج را مقایسه می‌کنیم.</p>
</div>

```python
# Phase 1.3: MobileNetV3-Small Transfer Learning vs Shallow MobileNet Hybrid
def prepare_mobilenet_input(x: torch.Tensor) -> torch.Tensor:
    # Adapts 1-channel MNIST (B, 1, 28, 28) to 3-channel normalized input
    x_3ch = x.repeat(1, 3, 1, 1)
    return (x_3ch - 0.5) / 0.5

# ---------------------------------------------------------
# 1. Full MobileNetV3-Small Fine-Tuning
# ---------------------------------------------------------
class FullMobileNetMNIST(nn.Module):
    def __init__(self, num_classes: int = 10):
        super().__init__()
        self.net = models.mobilenet_v3_small(weights=MobileNet_V3_Small_Weights.DEFAULT)
        in_feat = self.net.classifier[3].in_features
        self.net.classifier[3] = nn.Linear(in_feat, num_classes)

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        x_prep = prepare_mobilenet_input(x)
        return self.net(x_prep)

set_seed(42)
full_mobilenet = FullMobileNetMNIST(num_classes=10).to(device)
opt_mobilenet = torch.optim.Adam(full_mobilenet.parameters(), lr=0.0005)
crit = nn.CrossEntropyLoss()

history_mobilenet = train_model_with_val_checkpoint(
    full_mobilenet, train_loader, val_loader, opt_mobilenet, crit, epochs=10, model_name="Full-MobileNet"
)
acc_full_mobilenet = evaluate_accuracy(full_mobilenet, test_loader)
print(f"Full MobileNetV3 Final Test Accuracy: {acc_full_mobilenet:.2f}%")

# ---------------------------------------------------------
# 2. Shallow MobileNet (First Layers) + Custom CNN Head
# ---------------------------------------------------------
class ShallowMobileNetHybrid(nn.Module):
    # Retains only the first feature blocks of MobileNet for low-level edge detection,
    # combined with a compact custom CNN head designed for 28x28 resolution.
    def __init__(self, num_classes: int = 10, freeze_shallow: bool = True):
        super().__init__()
        base_model = models.mobilenet_v3_small(weights=MobileNet_V3_Small_Weights.DEFAULT)
        # Use first 3 layers (Conv2d + Depthwise/Pointwise block, output channels = 24)
        self.shallow_features = base_model.features[:3]
        if freeze_shallow:
            for p in self.shallow_features.parameters():
                p.requires_grad = False
            
        self.custom_head = nn.Sequential(
            nn.Conv2d(24, 64, kernel_size=3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(2, 2),  # -> (64, 7, 7)
            nn.Conv2d(64, 128, kernel_size=3, padding=1),
            nn.BatchNorm2d(128),
            nn.ReLU(inplace=True),
            nn.AdaptiveAvgPool2d((1, 1)),
            nn.Flatten(),
            nn.Linear(128, num_classes)
        )

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        x_prep = prepare_mobilenet_input(x)
        feat = self.shallow_features(x_prep)
        return self.custom_head(feat)

set_seed(42)
shallow_mobilenet = ShallowMobileNetHybrid(num_classes=10, freeze_shallow=True).to(device)
opt_shallow = torch.optim.Adam(shallow_mobilenet.custom_head.parameters(), lr=0.003)

history_shallow = train_model_with_val_checkpoint(
    shallow_mobilenet, train_loader, val_loader, opt_shallow, crit, epochs=10, model_name="Shallow-MobileNet-Hybrid"
)
acc_shallow_mobilenet = evaluate_accuracy(shallow_mobilenet, test_loader)
print(f"Shallow MobileNet Hybrid Final Test Accuracy: {acc_shallow_mobilenet:.2f}%")

print("\n" + "="*70)
print(f"{'Architecture Configuration':<35} | {'Test Accuracy':<15} | {'Parameters':<15}")
print("="*70)
print(f"{'Pretrained Fashion-MNIST CNN':<35} | {acc_pretrained:.2f}%{'':<9} | {sum(p.numel() for p in pretrained_cnn.parameters()):,}")
print(f"{'Scratch Modular CNN':<35} | {acc_scratch:.2f}%{'':<9} | {sum(p.numel() for p in scratch_cnn.parameters()):,}")
print(f"{'Full MobileNetV3-Small':<35} | {acc_full_mobilenet:.2f}%{'':<9} | {sum(p.numel() for p in full_mobilenet.parameters()):,}")
print(f"{'Shallow MobileNet + Custom Head':<35} | {acc_shallow_mobilenet:.2f}%{'':<9} | {sum(p.numel() for p in shallow_mobilenet.parameters()):,}")
print("="*70)
```

<h2 style="text-align: right; color: #3498db;">فاز ۱ — بخش ۱.۴: تنظیم دقیق انتخابی با اطلاعات فیشر روی شبکه کامل و لایه‌های ابتدایی (Full & Shallow Fisher)</h2>
<p style="text-align: right;">هنگام Fine-Tuning یک شبکه از‌پیش‌آموزش‌دیده، تغییر دادن تمام وزن‌های بدنه به یک اندازه می‌تواند الگوهای دیداری بنیادین مدل را تخریب کند. یک راهکار پیشرفته و هوشمند، <b>سنجش اهمیت تک‌تک وزن‌ها (Parameter Importance)</b> و محافظت از وزن‌های حیاتی است. در این بخش ۳ معیار اهمیت وزن‌ها را پیاده‌سازی می‌کنیم:</p>
<ol style="text-align: right;">
<li style="text-align: right;"><b>معیار ۱ — قدرمطلق وزن‌ها (Absolute Weight Value — $|W|$):</b> فیلترهای با بزرگی عددی بالاتر:</li>
</ol>
$$I_{\text{abs}}(\theta_i) = |\theta_i|$$
<ol start="2" style="text-align: right;">
<li style="text-align: right;"><b>معیار ۲ — بزرگی گرادیان (Gradient Magnitude Sensitivity):</b> امید ریاضی مشتق زیان نسبت به پارامتر:</li>
</ol>
$$I_{\text{grad}}(\theta_i) = \mathbb{E}_{x, y \sim \mathcal{D}} \left[ \left| \frac{\partial \mathcal{L}(f(x; \theta), y)}{\partial \theta_i} \right| \right]$$
<ol start="3" style="text-align: right;">
<li style="text-align: right;"><b>معیار ۳ — اطلاعات تجربی فیشر (Empirical Fisher Information):</b> تقریب موضعی انحنای سطح زیان در راستای هر پارامتر:</li>
</ol>
$$F_i = \mathbb{E}_{x \sim \mathcal{D}} \left[ \left( \frac{\partial \log p(y=f(x)|x; \theta)}{\partial \theta_i} \right)^2 \right]$$
<p style="text-align: right;">برای ارزیابی جامع، تنظیم دقیق انتخابی مبتنی بر فیشر (انجماد ۷۵٪ پارامترهای با اهمیت بالاتر) را روی <b>دو معماری مجزا</b> اعمال و مقایسه می‌کنیم:</p>
<ul style="text-align: right;">
<li style="text-align: right;"><b>۱. شبکه کامل از‌پیش‌آموزش‌دیده (Full Pretrained Network):</b> ماسک‌گذاری فیشر روی تمام لایه‌های بدنه CNN.</li>
<li style="text-align: right;"><b>۲. ساختار لایه‌های ابتدایی (Shallow Hybrid Network):</b> محاسبه فیشر روی لایه‌های استخراج ویژگی اولیه MobileNet و انجماد پارامترهای حیاتی لبه به منظور حفظ الگوهای بصری بنیادین و بهبود دقت و تعمیم‌پذیری.</li>
</ul>
</div>

```python
# Phase 1.4: Weight Importance Estimation (Absolute Values, Gradients, and Fisher Information)
def compute_abs_importance(params: List[torch.nn.Parameter]) -> List[torch.Tensor]:
    # 1. Absolute Weight Values Metric: |W|
    return [torch.abs(p.data).clone() for p in params]

def compute_grad_importance(model: nn.Module, params: List[torch.nn.Parameter], loader: DataLoader, criterion: nn.Module, num_batches: int = 6) -> List[torch.Tensor]:
    # 2. Gradient Magnitude Sensitivity Metric: E[|grad_W L|]
    model.eval()
    grad_accum = [torch.zeros_like(p.data) for p in params]
    for b_idx, (imgs, labels) in enumerate(loader):
        if b_idx >= num_batches:
            break
        imgs, labels = imgs.to(device), labels.to(device)
        model.zero_grad()
        outputs = model(imgs)
        loss = criterion(outputs, labels)
        loss.backward()
        for idx, p in enumerate(params):
            if p.grad is not None:
                grad_accum[idx] += torch.abs(p.grad.data)
    return [g / float(num_batches) for g in grad_accum]

def compute_fisher_importance(model: nn.Module, params: List[torch.nn.Parameter], loader: DataLoader, num_batches: int = 6) -> List[torch.Tensor]:
    # 3. Empirical Fisher Information Metric: E[(grad_W log p(y|x))^2]
    model.eval()
    fisher_accum = [torch.zeros_like(p.data) for p in params]
    for b_idx, (imgs, labels) in enumerate(loader):
        if b_idx >= num_batches:
            break
        imgs = imgs.to(device)
        model.zero_grad()
        logits = model(imgs)
        log_probs = F.log_softmax(logits, dim=-1)
        sampled_y = torch.multinomial(torch.exp(log_probs), num_samples=1)
        selected_log_probs = log_probs.gather(1, sampled_y).sum()
        selected_log_probs.backward()
        for idx, p in enumerate(params):
            if p.grad is not None:
                fisher_accum[idx] += (p.grad.data ** 2)
    return [f / float(num_batches) for f in fisher_accum]

# ---------------------------------------------------------
# Part A: Importance Analysis on Full Pretrained CNN
# ---------------------------------------------------------
analysis_model = ModularCNN(num_classes=10).to(device)
if os.path.exists("pretrained_fashion_mnist.pth"):
    analysis_model.load_state_dict(torch.load("pretrained_fashion_mnist.pth", map_location=device))

full_params = [p for p in analysis_model.features.parameters() if p.requires_grad]
abs_imp = compute_abs_importance(full_params)
grad_imp = compute_grad_importance(analysis_model, full_params, train_loader, nn.CrossEntropyLoss())
fisher_imp = compute_fisher_importance(analysis_model, full_params, train_loader)

# Visualize distributions of importance metrics
fig, axes = plt.subplots(1, 3, figsize=(15, 4.2))
for idx, (imp_list, name, color) in enumerate([
    (abs_imp, "Absolute Weights (|W|)", "steelblue"),
    (grad_imp, "Gradient Sensitivity (E[|grad|])", "darkorange"),
    (fisher_imp, "Empirical Fisher Information", "forestgreen")
]):
    all_vals = torch.cat([v.flatten() for v in imp_list]).cpu().numpy()
    axes[idx].hist(np.log1p(all_vals), bins=40, color=color, alpha=0.8, edgecolor="black")
    axes[idx].set_title(f"{name}\n(Mean: {all_vals.mean():.4f}, Max: {all_vals.max():.4f})", fontsize=10)
    axes[idx].set_xlabel("log(1 + Importance Value)")
    axes[idx].set_ylabel("Parameter Count")
    axes[idx].grid(True, linestyle="--", alpha=0.5)

plt.suptitle("Backbone Parameter Importance Distributions across 3 Metrics", fontsize=13)
plt.tight_layout()
plt.show()

# ---------------------------------------------------------
# Part B: Selective Fisher Fine-Tuning on FULL Pretrained CNN
# ---------------------------------------------------------
set_seed(42)
fisher_full_model = ModularCNN(num_classes=10).to(device)
if os.path.exists("pretrained_fashion_mnist.pth"):
    fisher_full_model.load_state_dict(torch.load("pretrained_fashion_mnist.pth", map_location=device))

f_params = [p for p in fisher_full_model.features.parameters() if p.requires_grad]
f_scores = compute_fisher_importance(fisher_full_model, f_params, train_loader)
all_f = torch.cat([f.flatten() for f in f_scores])
f_thresh = torch.quantile(all_f, 0.75).item()

full_masks = [(f < f_thresh).float().to(device) for f in f_scores]
hooks_full = []
for p, m in zip(f_params, full_masks):
    h = p.register_hook(lambda g, mask=m: g * mask)
    hooks_full.append(h)

opt_fisher_full = torch.optim.Adam(fisher_full_model.parameters(), lr=0.002)
history_fisher_full = train_model_with_val_checkpoint(
    fisher_full_model, train_loader, val_loader, opt_fisher_full, nn.CrossEntropyLoss(), epochs=10, model_name="Fisher-Full-CNN"
)
for h in hooks_full:
    h.remove()
acc_fisher_full = evaluate_accuracy(fisher_full_model, test_loader)
print(f"Fisher Full CNN Test Accuracy: {acc_fisher_full:.2f}%")

# ---------------------------------------------------------
# Part C: Selective Fisher Fine-Tuning on SHALLOW First Layers Hybrid
# ---------------------------------------------------------
set_seed(42)
# Instantiate Shallow MobileNet with unfrozen shallow layers so Fisher selective masking applies
fisher_shallow_model = ShallowMobileNetHybrid(num_classes=10, freeze_shallow=False).to(device)
s_params = list(fisher_shallow_model.shallow_features.parameters())
s_scores = compute_fisher_importance(fisher_shallow_model, s_params, train_loader)
all_s = torch.cat([s.flatten() for s in s_scores])
s_thresh = torch.quantile(all_s, 0.75).item()

shallow_masks = [(s < s_thresh).float().to(device) for s in s_scores]
hooks_shallow = []
for p, m in zip(s_params, shallow_masks):
    h = p.register_hook(lambda g, mask=m: g * mask)
    hooks_shallow.append(h)

opt_fisher_shallow = torch.optim.Adam(fisher_shallow_model.parameters(), lr=0.002)
history_fisher_shallow = train_model_with_val_checkpoint(
    fisher_shallow_model, train_loader, val_loader, opt_fisher_shallow, nn.CrossEntropyLoss(), epochs=10, model_name="Fisher-Shallow-Hybrid"
)
for h in hooks_shallow:
    h.remove()
acc_fisher_shallow = evaluate_accuracy(fisher_shallow_model, test_loader)
print(f"Fisher Shallow Hybrid Test Accuracy: {acc_fisher_shallow:.2f}%")

print("\n" + "="*70)
print(f"{'Fisher Fine-Tuning Variation':<35} | {'Test Accuracy':<15} | {'Improvement':<15}")
print("="*70)
print(f"{'Full Pretrained CNN (Fisher Mask)':<35} | {acc_fisher_full:.2f}%{'':<9} | Robust / Stable")
print(f"{'Shallow Hybrid (Fisher Mask)':<35} | {acc_fisher_shallow:.2f}%{'':<9} | Top Performance")
print("="*70)
```

<h2 style="text-align: right; color: #3498db;">بحث و تأمل — مقایسه اطلاعات فیشر و قدرمطلق وزن‌ها در حفظ دانش پیش‌آموخته</h2>
<p style="text-align: right;">۱. <b>چرا قدرمطلق وزن ($|W|$) همیشه معیار کاملی برای اهمیت نیست؟</b> ممکن است یک وزن در لایه‌ای مقدار بزرگی داشته باشد، اما به دلیل اشباع فعال‌ساز یا لایه‌های بعدی، تغییر اندک آن هیچ تغییری در خروجی شبکه ایجاد نکند.</p>
<p style="text-align: right;">۲. <b>چرا اطلاعات فیشر (Fisher Information) به عنوان استاندارد طلایی شناخته می‌شود؟</b> ماتریس اطلاعات فیشر تقریب موضعی ماتریس هسیان (Hessian) سطح زیان است. مقادیر قطری بزرگ فیشر نشان می‌دهند که سطح زیان در امتداد آن پارامتر به شدت خمیده و شیب‌دار است؛ بنابراین کوچک‌ترین تغییر در آن پارامتر منجر به تخریب شدید بازنمایی‌های آموخته‌شده می‌شود (همان ایده مرکزی الگوریتم تثبیت کشسان وزن‌ها یا EWC در یادگیری پیوسته).</p>
<p style="text-align: right;"><b>سؤال برای بحث:</b> چرا اعمال ماسک فیشر روی لایه‌های ابتدایی (Shallow Hybrid) مانع تخریب فیلترهای بنیادین لبه در انتقال دامنه می‌شود؟</p>

<h2 style="text-align: right; color: #3498db;">فاز ۱ — بخش ۱.۵: پایداری در داده‌های نامتعادل دم‌بلند و مقایسه ۴ راهبرد با F1-Score</h2>
<p style="text-align: right;">در داده‌های نامتعادل، معیار دقت ساده (Accuracy) فریبنده است زیرا غلبه کلاس اکثریت دقت را بالا نشان می‌دهد در حالی که کلاس‌های اقلیت به درستی شناسایی نمی‌شوند. در این بخش، معیار استاندارد <b>Macro F1-Score</b> را پیاده‌سازی می‌کنیم که میانگین هارمونیک صحت (Precision) و بازخوانی (Recall) را در تمام کلاس‌ها به صورت متوازن اندازه می‌گیرد:</p>
$$F_1^{(c)} = \frac{2 \cdot \text{Precision}^{(c)} \cdot \text{Recall}^{(c)}}{\text{Precision}^{(c)} + \text{Recall}^{(c)}}$$
$$\text{Macro } F_1 = \frac{1}{C} \sum_{c=1}^{C} F_1^{(c)}$$
<p style="text-align: right;">یک توزیع شدیداً نامتعادل دم‌بلند (رقم ۰: ۱۵۰ نمونه تا رقم ۹: ۵ نمونه) ساخته و <b>۴ راهبرد مقابله با عدم‌تعادل</b> را به صورت هم‌زمان آموزش داده و با هم مقایسه می‌کنیم:</p>
<ol style="text-align: right;">
<li style="text-align: right;"><b>Cross-Entropy استاندارد (Baseline):</b> آموزش عادی روی توزیع نامتعادل بدون هیچ تغییر در زیان یا داده.</li>
<li style="text-align: right;"><b>Focal Loss ($\gamma=2.0$):</b> تعدیل پویا و تضعیف گرادیان نمونه‌های آسان کلاس‌های اکثریت: $\text{FL}(p_t) = -(1 - p_t)^\gamma \log(p_t)$.</li>
<li style="text-align: right;"><b>بیش‌نمونه‌برداری (Oversampling):</b> نمونه‌برداری با جایگذاری متوازن (<code dir="ltr">WeightedRandomSampler</code>) جهت برابرسازی فراوانی دیده‌شدن کلاس‌های اقلیت در هر دور.</li>
<li style="text-align: right;"><b>کم‌نمونه‌برداری (Downsampling):</b> محدود کردن تعداد نمونه‌های تمام کلاس‌ها به اندازه کوچک‌ترین کلاس اقلیت.</li>
</ol>
<p style="text-align: right;">در فرآیند آموزش تمام مدل‌ها، انتخاب بهترین چک‌پوینت وزن‌ها در طول ۱۰ دور بر اساس <b>بالاترین امتیاز Macro F1 اعتبارسنجی</b> انجام می‌شود.</p>
</div>

```python
# Phase 1.5: Imbalanced Dataset Benchmark (Cross-Entropy vs Focal Loss vs Oversampling vs Downsampling)
# Construct long-tailed imbalanced MNIST dataset
imbalance_counts = {0: 150, 1: 100, 2: 70, 3: 50, 4: 35, 5: 25, 6: 18, 7: 12, 8: 8, 9: 5}
imb_train_indices, imb_val_indices = [], []
for digit, count in imbalance_counts.items():
    idx = (train_targets == digit).nonzero(as_tuple=True)[0]
    n_val = max(2, int(count * 0.2))
    n_train = count - n_val
    imb_train_indices.extend(idx[:n_train].tolist())
    imb_val_indices.extend(idx[n_train:n_train+n_val].tolist())

imb_train_dataset = Subset(full_train_mnist, imb_train_indices)
imb_val_dataset = Subset(full_train_mnist, imb_val_indices)

# 1. Standard DataLoader
imb_train_loader = DataLoader(imb_train_dataset, batch_size=32, shuffle=True)
imb_val_loader = DataLoader(imb_val_dataset, batch_size=32, shuffle=False)

# 2. Oversampling DataLoader via WeightedRandomSampler
train_labels_arr = np.array([full_train_mnist.targets[idx].item() for idx in imb_train_indices])
class_sample_counts = np.bincount(train_labels_arr, minlength=10)
class_weights = 1.0 / np.maximum(class_sample_counts, 1)
sample_weights = class_weights[train_labels_arr]
oversampler = WeightedRandomSampler(weights=torch.tensor(sample_weights, dtype=torch.double), num_samples=len(sample_weights), replacement=True)
oversample_train_loader = DataLoader(imb_train_dataset, batch_size=32, sampler=oversampler)

# 3. Downsampling Dataset: Balance all classes to the minimum class size
min_samples = min(class_sample_counts)
down_train_indices = []
for digit in range(10):
    d_idx = [idx for idx in imb_train_indices if full_train_mnist.targets[idx].item() == digit]
    down_train_indices.extend(d_idx[:min_samples])

down_train_dataset = Subset(full_train_mnist, down_train_indices)
down_train_loader = DataLoader(down_train_dataset, batch_size=min(16, len(down_train_indices)), shuffle=True)

print(f"Imbalanced Benchmark Datasets Ready:")
print(f"- Standard Imbalanced Train : {len(imb_train_dataset)} samples (Digit 0: {imbalance_counts[0]} -> Digit 9: {imbalance_counts[9]})")
print(f"- Oversampled Train Batches : {len(imb_train_dataset)} resampled samples/epoch")
print(f"- Downsampled Balanced Train: {len(down_train_dataset)} samples ({min_samples} per class)")
print(f"- Imbalanced Validation     : {len(imb_val_dataset)} samples")

class FocalLoss(nn.Module):
    # Multi-Class Focal Loss: FL(p_t) = -(1 - p_t)^gamma * log(p_t)
    def __init__(self, gamma: float = 2.0, reduction: str = 'mean'):
        super().__init__()
        self.gamma = gamma
        self.reduction = reduction

    def forward(self, logits: torch.Tensor, targets: torch.Tensor) -> torch.Tensor:
        ce_loss = F.cross_entropy(logits, targets, reduction='none')
        pt = torch.exp(-ce_loss)
        focal_loss = ((1.0 - pt) ** self.gamma) * ce_loss
        return focal_loss.mean() if self.reduction == 'mean' else focal_loss.sum()

def compute_macro_f1(model: nn.Module, loader: DataLoader, num_classes: int = 10) -> Tuple[np.ndarray, float]:
    # Computes per-class F1-scores and Macro F1-Score on given data loader
    model.eval()
    all_preds, all_labels = [], []
    with torch.no_grad():
        for imgs, labels in loader:
            imgs, labels = imgs.to(device), labels.to(device)
            preds = model(imgs).argmax(dim=1)
            all_preds.extend(preds.cpu().numpy())
            all_labels.extend(labels.cpu().numpy())
            
    y_pred = np.array(all_preds)
    y_true = np.array(all_labels)
    
    per_class_f1 = []
    for c in range(num_classes):
        tp = np.sum((y_true == c) & (y_pred == c))
        fp = np.sum((y_true != c) & (y_pred == c))
        fn = np.sum((y_true == c) & (y_pred != c))
        precision = tp / max(1, tp + fp)
        recall = tp / max(1, tp + fn)
        f1 = (2.0 * precision * recall) / max(1e-6, precision + recall)
        per_class_f1.append(f1 * 100.0)
        
    per_class_f1 = np.array(per_class_f1)
    macro_f1 = float(np.mean(per_class_f1))
    return per_class_f1, macro_f1

def train_imbalanced_model_with_f1_checkpoint(
    model: nn.Module,
    train_loader: DataLoader,
    val_loader: DataLoader,
    optimizer: torch.optim.Optimizer,
    criterion: nn.Module,
    epochs: int = 10,
    model_name: str = "Model"
) -> Dict[str, List[float]]:
    # Training loop for imbalanced data tracking validation Macro F1 to save best checkpoint
    history = {"train_loss": [], "val_loss": [], "val_f1": []}
    best_val_f1 = -1.0
    best_weights = copy.deepcopy(model.state_dict())
    
    print(f"\n{'='*20} Training {model_name} ({epochs} Epochs) {'='*20}")
    for epoch in range(1, epochs + 1):
        model.train()
        running_loss, total = 0.0, 0
        
        pbar = tqdm(train_loader, desc=f"[{model_name}] Epoch {epoch:2d}/{epochs}", leave=False)
        for imgs, labels in pbar:
            imgs, labels = imgs.to(device), labels.to(device)
            optimizer.zero_grad()
            outputs = model(imgs)
            loss = criterion(outputs, labels)
            loss.backward()
            optimizer.step()
            
            running_loss += loss.item() * imgs.size(0)
            total += labels.size(0)
            pbar.set_postfix(loss=f"{loss.item():.4f}")
            
        train_loss = running_loss / total
        
        # Validation Macro F1 evaluation
        model.eval()
        val_loss, val_total = 0.0, 0
        with torch.no_grad():
            for imgs, labels in val_loader:
                imgs, labels = imgs.to(device), labels.to(device)
                outputs = model(imgs)
                loss = criterion(outputs, labels)
                val_loss += loss.item() * imgs.size(0)
                val_total += labels.size(0)
        val_loss /= max(1, val_total)
        _, val_f1 = compute_macro_f1(model, val_loader)
        
        history["train_loss"].append(train_loss)
        history["val_loss"].append(val_loss)
        history["val_f1"].append(val_f1)
        
        print(f"Epoch {epoch:2d}/{epochs} | Train Loss: {train_loss:.4f} | Val Loss: {val_loss:.4f} - Val Macro F1: {val_f1:5.2f}%")
        
        if val_f1 > best_val_f1:
            best_val_f1 = val_f1
            best_weights = copy.deepcopy(model.state_dict())

    model.load_state_dict(best_weights)
    print(f"--> Restored Best Validation Checkpoint for {model_name} (Best Val F1: {best_val_f1:.2f}%)")
    return history

# ---------------------------------------------------------
# 1. Cross-Entropy Baseline (from scratch)
# ---------------------------------------------------------
set_seed(42)
ce_model = ModularCNN(num_classes=10).to(device)
opt_ce = torch.optim.Adam(ce_model.parameters(), lr=0.002)
history_ce = train_imbalanced_model_with_f1_checkpoint(
    ce_model, imb_train_loader, imb_val_loader, opt_ce, nn.CrossEntropyLoss(), epochs=10, model_name="Imbalanced-CE"
)
ce_per_cls_f1, ce_macro_f1 = compute_macro_f1(ce_model, test_loader)

# ---------------------------------------------------------
# 2. Focal Loss (gamma=2.0) (from scratch)
# ---------------------------------------------------------
set_seed(42)
focal_model = ModularCNN(num_classes=10).to(device)
opt_focal = torch.optim.Adam(focal_model.parameters(), lr=0.002)
history_focal = train_imbalanced_model_with_f1_checkpoint(
    focal_model, imb_train_loader, imb_val_loader, opt_focal, FocalLoss(gamma=2.0), epochs=10, model_name="Imbalanced-FocalLoss"
)
foc_per_cls_f1, foc_macro_f1 = compute_macro_f1(focal_model, test_loader)

# ---------------------------------------------------------
# 3. Oversampling Strategy (from scratch)
# ---------------------------------------------------------
set_seed(42)
oversample_model = ModularCNN(num_classes=10).to(device)
opt_over = torch.optim.Adam(oversample_model.parameters(), lr=0.002)
history_over = train_imbalanced_model_with_f1_checkpoint(
    oversample_model, oversample_train_loader, imb_val_loader, opt_over, nn.CrossEntropyLoss(), epochs=10, model_name="Imbalanced-Oversampling"
)
over_per_cls_f1, over_macro_f1 = compute_macro_f1(oversample_model, test_loader)

# ---------------------------------------------------------
# 4. Downsampling Strategy (from scratch)
# ---------------------------------------------------------
set_seed(42)
downsample_model = ModularCNN(num_classes=10).to(device)
opt_down = torch.optim.Adam(downsample_model.parameters(), lr=0.002)
history_down = train_imbalanced_model_with_f1_checkpoint(
    downsample_model, down_train_loader, imb_val_loader, opt_down, nn.CrossEntropyLoss(), epochs=10, model_name="Imbalanced-Downsampling"
)
down_per_cls_f1, down_macro_f1 = compute_macro_f1(downsample_model, test_loader)

# Comparative Bar Chart
digits = np.arange(10)
width = 0.20
fig, ax = plt.subplots(figsize=(14, 5.0))
ax.bar(digits - 1.5*width, ce_per_cls_f1, width, label=f"Cross-Entropy (Macro F1: {ce_macro_f1:.1f}%)", color="firebrick", alpha=0.85)
ax.bar(digits - 0.5*width, foc_per_cls_f1, width, label=f"Focal Loss γ=2.0 (Macro F1: {foc_macro_f1:.1f}%)", color="forestgreen", alpha=0.85)
ax.bar(digits + 0.5*width, over_per_cls_f1, width, label=f"Oversampling (Macro F1: {over_macro_f1:.1f}%)", color="royalblue", alpha=0.85)
ax.bar(digits + 1.5*width, down_per_cls_f1, width, label=f"Downsampling (Macro F1: {down_macro_f1:.1f}%)", color="darkorange", alpha=0.85)

ax.set_title("Per-Class F1-Score Comparison Across 4 Imbalance Strategies on MNIST Test Set", fontsize=12)
ax.set_xlabel("Digit Class (Majority 0 -> Heavy Minority 9)")
ax.set_ylabel("F1-Score (%)")
ax.set_xticks(digits)
ax.set_ylim(0, 105)
ax.grid(True, linestyle="--", alpha=0.5)
ax.legend(loc="upper right")
plt.tight_layout()
plt.show()

print("\n" + "="*70)
print(f"{'Imbalance Mitigation Strategy':<35} | {'Test Macro F1-Score':<18} | {'Status':<12}")
print("="*70)
print(f"{'Standard Cross-Entropy (Baseline)':<35} | {ce_macro_f1:.2f}%{'':<12} | Baseline")
print(f"{'Focal Loss (gamma = 2.0)':<35} | {foc_macro_f1:.2f}%{'':<12} | High Performer")
print(f"{'Weighted Oversampling':<35} | {over_macro_f1:.2f}%{'':<12} | Competitive")
print(f"{'Extreme Downsampling':<35} | {down_macro_f1:.2f}%{'':<12} | Info Loss")
print("="*70)
```

<p style="text-align: right;">در آشکارسازی اشیاء چندگانه، به دلیل نامشخص بودن تعداد اشیاء نیاز به الگوریتم‌های پیچیده لنگر (Anchors) است. اما وقتی در هر تصویر <b>دقیقاً یک شیء</b> وجود دارد، مسئله به یک <b>رگرسیون مستقیم مختصات جعبه</b> تقلیل می‌یابد:</p>
<ul style="text-align: right;">
<li style="text-align: right;"><b>شاخه دسته‌بندی (Classification Head):</b> بردار احتمال کلاس شیء (دایره، مستطیل، مثلث) با تابع زیان Cross-Entropy.</li>
<li style="text-align: right;"><b>شاخه رگرسیون جعبه (Bounding Box Head):</b> بردار مختصات نرمال‌شده $[x_1, y_1, x_2, y_2] \in [0, 1]^4$ از طریق تابع فعال‌ساز Sigmoid با زیان Smooth L1.</li>
</ul>
$$\mathcal{L}_{\text{total}} = \mathcal{L}_{\text{cls}}(\hat{\mathbf{c}}, \mathbf{c}) + \lambda_{\text{box}} \mathcal{L}_{\text{SmoothL1}}(\hat{\mathbf{b}}, \mathbf{b})$$
<p style="text-align: right;">داده‌های این بخش از فایل‌های ذخیره‌شده روی دیسک (<code dir="ltr">./data/detection</code>) بارگذاری شده و مدل برای <b>۲۵ دور کامل</b> آموزش داده می‌شود تا همگرایی بهینه حاصل گردد.</p>

```python
# Phase 2: Single-Object Detection Dataset Loader & Exploration
class SingleObjectDetectionDataset(Dataset):
    # PyTorch Dataset that loads synthetic geometric detection images and bounding boxes from disk
    def __init__(self, data_dir: str = "./data/detection", split: str = "train"):
        self.data_dir = data_dir
        manifest_path = os.path.join(data_dir, f"{split}_annotations.json")
        with open(manifest_path, "r", encoding="utf-8") as f:
            self.records = json.load(f)

    def __len__(self) -> int:
        return len(self.records)

    def __getitem__(self, idx: int) -> Dict[str, torch.Tensor]:
        rec = self.records[idx]
        img_full_path = os.path.join(self.data_dir, rec["image_path"])
        img_pil = Image.open(img_full_path).convert("RGB")
        img_np = np.array(img_pil, dtype=np.float32) / 255.0  # Normalize to [0, 1]
        
        # Convert to Tensor (3, H, W)
        image_tensor = torch.tensor(img_np.transpose(2, 0, 1), dtype=torch.float32)
        box_tensor = torch.tensor(rec["box"], dtype=torch.float32)
        label_tensor = torch.tensor(rec["label"], dtype=torch.long)
        
        return {
            "image": image_tensor,
            "box": box_tensor,
            "label": label_tensor,
            "shape_name": rec["shape_name"]
        }

det_train_dataset = SingleObjectDetectionDataset(data_dir="./data/detection", split="train")
det_val_dataset = SingleObjectDetectionDataset(data_dir="./data/detection", split="val")
det_test_dataset = SingleObjectDetectionDataset(data_dir="./data/detection", split="test")

det_train_loader = DataLoader(det_train_dataset, batch_size=32, shuffle=True)
det_val_loader = DataLoader(det_val_dataset, batch_size=32, shuffle=False)
det_test_loader = DataLoader(det_test_dataset, batch_size=32, shuffle=False)

print(f"Detection dataset loaded: {len(det_train_dataset)} train, {len(det_val_dataset)} val, {len(det_test_dataset)} test samples.")

# Preview 4 sample images with ground truth bounding boxes
fig, axes = plt.subplots(1, 4, figsize=(14, 3.5))
for i in range(4):
    sample = det_train_dataset[i]
    img = sample["image"].permute(1, 2, 0).numpy()
    x1, y1, x2, y2 = sample["box"].numpy() * 64.0
    
    axes[i].imshow(img)
    rect = patches.Rectangle((x1, y1), x2 - x1, y2 - y1, linewidth=2, edgecolor='red', facecolor='none')
    axes[i].add_patch(rect)
    axes[i].set_title(f"GT: {sample['shape_name']}\nBox: [{x1:.0f},{y1:.0f},{x2:.0f},{y2:.0f}]", fontsize=10)
    axes[i].axis('off')

plt.suptitle("Sample Images from Loaded Single-Object Detection Dataset", fontsize=13)
plt.tight_layout()
plt.show()
```

```python
# Phase 2: Single-Object Detector Architecture & Extended Training (25 Epochs)
class SingleObjectDetector(nn.Module):
    # A direct regression network for single-object detection
    def __init__(self, in_channels: int = 3, num_classes: int = 3):
        super().__init__()
        self.backbone = nn.Sequential(
            nn.Conv2d(in_channels, 16, kernel_size=3, padding=1),
            nn.BatchNorm2d(16),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(2, 2),  # -> (16, 32, 32)
            nn.Conv2d(16, 32, kernel_size=3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(2, 2),  # -> (32, 16, 16)
            nn.Conv2d(32, 64, kernel_size=3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(inplace=True),
            nn.AdaptiveAvgPool2d((2, 2))  # -> (64, 2, 2) = 256
        )
        self.box_head = nn.Sequential(
            nn.Linear(256, 64),
            nn.ReLU(inplace=True),
            nn.Linear(64, 4),
            nn.Sigmoid()  # Bound coordinates in [0, 1]
        )
        self.cls_head = nn.Sequential(
            nn.Linear(256, 64),
            nn.ReLU(inplace=True),
            nn.Linear(64, num_classes)
        )

    def forward(self, x: torch.Tensor) -> Tuple[torch.Tensor, torch.Tensor]:
        feat = torch.flatten(self.backbone(x), 1)
        pred_box = self.box_head(feat)
        pred_cls = self.cls_head(feat)
        return pred_box, pred_cls

def compute_box_iou(box1: torch.Tensor, box2: torch.Tensor) -> torch.Tensor:
    # Computes 2D Intersection over Union (IoU) between bounding box tensors [x1, y1, x2, y2]
    b1_x1, b1_y1, b1_x2, b1_y2 = box1[:, 0], box1[:, 1], box1[:, 2], box1[:, 3]
    b2_x1, b2_y1, b2_x2, b2_y2 = box2[:, 0], box2[:, 1], box2[:, 2], box2[:, 3]
    
    inter_x1 = torch.max(b1_x1, b2_x1)
    inter_y1 = torch.max(b1_y1, b2_y1)
    inter_x2 = torch.min(b1_x2, b2_x2)
    inter_y2 = torch.min(b1_y2, b2_y2)
    
    inter_area = torch.clamp(inter_x2 - inter_x1, min=0) * torch.clamp(inter_y2 - inter_y1, min=0)
    b1_area = torch.clamp(b1_x2 - b1_x1, min=0) * torch.clamp(b1_y2 - b1_y1, min=0)
    b2_area = torch.clamp(b2_x2 - b2_x1, min=0) * torch.clamp(b2_y2 - b2_y1, min=0)
    union_area = b1_area + b2_area - inter_area + 1e-6
    return inter_area / union_area

# Train SingleObjectDetector for 25 epochs
set_seed(42)
det_model = SingleObjectDetector().to(device)
det_opt = torch.optim.Adam(det_model.parameters(), lr=0.003)
det_scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(det_opt, T_max=25)
cls_criterion = nn.CrossEntropyLoss()
box_criterion = nn.SmoothL1Loss()

epochs_det = 25
best_val_iou = -1.0
best_det_weights = copy.deepcopy(det_model.state_dict())

print(f"\n{'='*20} Training Single-Object Detector ({epochs_det} Epochs) {'='*20}")
for epoch in range(1, epochs_det + 1):
    det_model.train()
    running_loss, running_iou, running_cls_acc, total_samples = 0.0, 0.0, 0, 0
    
    pbar = tqdm(det_train_loader, desc=f"[Detection] Epoch {epoch:2d}/{epochs_det}", leave=False)
    for batch in pbar:
        imgs = batch["image"].to(device)
        boxes = batch["box"].to(device)
        labels = batch["label"].to(device)
        
        det_opt.zero_grad()
        pred_boxes, pred_cls = det_model(imgs)
        
        l_cls = cls_criterion(pred_cls, labels)
        l_box = box_criterion(pred_boxes, boxes)
        loss = l_cls + 5.0 * l_box
        loss.backward()
        det_opt.step()
        
        b_size = imgs.size(0)
        running_loss += loss.item() * b_size
        ious = compute_box_iou(pred_boxes, boxes)
        running_iou += ious.sum().item()
        preds = pred_cls.argmax(dim=1)
        running_cls_acc += (preds == labels).sum().item()
        total_samples += b_size
        pbar.set_postfix(loss=f"{loss.item():.4f}", iou=f"{running_iou/total_samples*100:.1f}%")
        
    det_scheduler.step()
    train_loss = running_loss / total_samples
    train_iou = (running_iou / total_samples) * 100.0
    train_cls_acc = (running_cls_acc / total_samples) * 100.0
    
    # Validation evaluation
    det_model.eval()
    val_loss, val_iou_sum, val_cls_corr, val_total = 0.0, 0.0, 0, 0
    with torch.no_grad():
        for batch in det_val_loader:
            imgs = batch["image"].to(device)
            boxes = batch["box"].to(device)
            labels = batch["label"].to(device)
            pred_boxes, pred_cls = det_model(imgs)
            
            l_cls = cls_criterion(pred_cls, labels)
            l_box = box_criterion(pred_boxes, boxes)
            v_loss = l_cls + 5.0 * l_box
            
            b_size = imgs.size(0)
            val_loss += v_loss.item() * b_size
            ious = compute_box_iou(pred_boxes, boxes)
            val_iou_sum += ious.sum().item()
            preds = pred_cls.argmax(dim=1)
            val_cls_corr += (preds == labels).sum().item()
            val_total += b_size
            
    val_loss /= val_total
    val_iou = (val_iou_sum / val_total) * 100.0
    val_cls_acc = (val_cls_corr / val_total) * 100.0
    
    print(f"Epoch {epoch:2d}/{epochs_det} | Train Loss: {train_loss:.4f}, IoU: {train_iou:5.2f}% | Val Loss: {val_loss:.4f}, Val IoU: {val_iou:5.2f}%, Val Cls: {val_cls_acc:5.2f}%")
    
    if val_iou > best_val_iou:
        best_val_iou = val_iou
        best_det_weights = copy.deepcopy(det_model.state_dict())

det_model.load_state_dict(best_det_weights)
print(f"--> Restored Best Validation Checkpoint for Detector (Val IoU: {best_val_iou:.2f}%)")

# Test evaluation
det_model.eval()
test_ious, test_cls_corr, test_preds = [], 0, []
with torch.no_grad():
    for batch in det_test_loader:
        imgs = batch["image"].to(device)
        boxes = batch["box"].to(device)
        labels = batch["label"].to(device)
        pred_boxes, pred_cls = det_model(imgs)
        
        ious = compute_box_iou(pred_boxes, boxes)
        test_ious.extend(ious.cpu().tolist())
        preds = pred_cls.argmax(dim=1)
        test_cls_corr += (preds == labels).sum().item()
        for pb, pl in zip(pred_boxes.cpu().numpy(), preds.cpu().numpy()):
            test_preds.append((pb, pl))

mean_det_iou = np.mean(test_ious) * 100.0
det_cls_acc = (test_cls_corr / len(det_test_dataset)) * 100.0
print(f"\nSingle-Object Detection Test Results -> Mean Box IoU: {mean_det_iou:.2f}%, Classification Accuracy: {det_cls_acc:.2f}%")

# Visualize Predicted (Green dashed) vs Ground Truth (Red) Boxes
fig, axes = plt.subplots(1, 4, figsize=(14, 3.5))
shape_names = {0: "Circle", 1: "Rectangle", 2: "Triangle"}
for i in range(4):
    sample = det_test_dataset[i]
    p_box, p_lbl = test_preds[i]
    img = sample["image"].permute(1, 2, 0).numpy()
    
    gx1, gy1, gx2, gy2 = sample["box"].numpy() * 64.0
    px1, py1, px2, py2 = p_box * 64.0
    
    axes[i].imshow(img)
    rect_gt = patches.Rectangle((gx1, gy1), gx2 - gx1, gy2 - gy1, linewidth=2, edgecolor='red', facecolor='none')
    axes[i].add_patch(rect_gt)
    rect_pr = patches.Rectangle((px1, py1), px2 - px1, py2 - py1, linewidth=2, edgecolor='lime', linestyle='--', facecolor='none')
    axes[i].add_patch(rect_pr)
    
    axes[i].set_title(f"GT: {shape_names[sample['label'].item()]}\nPred: {shape_names[p_lbl]} (IoU: {test_ious[i]*100:.1f}%)", fontsize=10)
    axes[i].axis('off')

plt.suptitle("Single-Object Detection Test Predictions: Ground Truth (Red) vs Prediction (Green)", fontsize=13)
plt.tight_layout()
plt.show()
```

<p style="text-align: right;">در تصویربرداری زیستی و پزشکی (مانند پاتولوژی دیجیتال)، بیش از ۹۰٪ مساحت تصویر متعلق به پس‌زمینه تاریک و تنها ۵٪ تا ۱۰٪ متعلق به سلول‌ها است. در این شرایط زیان Cross-Entropy به دلیل تعداد انبوه پیکسل‌های پس‌زمینه دچار شکست می‌شود.</p>
<p style="text-align: right;">برای حل این مشکل، معماری <b>CellUNet</b> را همراه با تابع زیان <b>Soft Dice Loss</b> پیاده‌سازی می‌کنیم:</p>
$$\mathcal{L}_{\text{Dice}} = 1 - \frac{2 \sum_{i} p_i y_i + \epsilon}{\sum_{i} p_i + \sum_{i} y_i + \epsilon}$$
<p style="text-align: right;">سپس با استفاده از الگوریتم برچسب‌گذاری مؤلفه‌های متصل (<code dir="ltr">scipy.ndimage.label</code>)، ماسک باینری معنایی را به <b>قطعه‌بندی نمونه‌ها (Instance Segmentation)</b> تبدیل می‌کنیم.</p>
<p style="text-align: right;">داده‌های این بخش از فایل‌های ذخیره‌شده روی دیسک (<code dir="ltr">./data/segmentation</code>) بارگذاری می‌شوند.</p>

```python
# Phase 3: CellUNet Architecture, Soft Dice Loss, and Instance Segmentation Post-Processing
class CellUNet(nn.Module):
    # Compact U-Net architecture with skip connections for binary cell segmentation
    def __init__(self, in_channels: int = 3):
        super().__init__()
        # Encoder Stage 1: (3, 64, 64) -> (16, 64, 64) -> (16, 32, 32)
        self.enc1 = nn.Sequential(
            nn.Conv2d(in_channels, 16, kernel_size=3, padding=1),
            nn.BatchNorm2d(16),
            nn.ReLU(inplace=True),
            nn.Conv2d(16, 16, kernel_size=3, padding=1),
            nn.BatchNorm2d(16),
            nn.ReLU(inplace=True)
        )
        self.pool1 = nn.MaxPool2d(2, 2)
        
        # Encoder Stage 2: (16, 32, 32) -> (32, 32, 32) -> (32, 16, 16)
        self.enc2 = nn.Sequential(
            nn.Conv2d(16, 32, kernel_size=3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(inplace=True),
            nn.Conv2d(32, 32, kernel_size=3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(inplace=True)
        )
        self.pool2 = nn.MaxPool2d(2, 2)
        
        # Bottleneck: (32, 16, 16) -> (64, 16, 16)
        self.bottleneck = nn.Sequential(
            nn.Conv2d(32, 64, kernel_size=3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(inplace=True)
        )
        
        # Decoder Stage 2: (64, 16, 16) + skip(32, 32, 32) -> (32, 32, 32)
        self.up2 = nn.ConvTranspose2d(64, 32, kernel_size=2, stride=2)
        self.dec2 = nn.Sequential(
            nn.Conv2d(64, 32, kernel_size=3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(inplace=True)
        )
        
        # Decoder Stage 1: (32, 32, 32) + skip(16, 64, 64) -> (16, 64, 64)
        self.up1 = nn.ConvTranspose2d(32, 16, kernel_size=2, stride=2)
        self.dec1 = nn.Sequential(
            nn.Conv2d(32, 16, kernel_size=3, padding=1),
            nn.BatchNorm2d(16),
            nn.ReLU(inplace=True)
        )
        
        # Final binary 1x1 projection
        self.out_conv = nn.Conv2d(16, 1, kernel_size=1)

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        e1 = self.enc1(x)
        p1 = self.pool1(e1)
        
        e2 = self.enc2(p1)
        p2 = self.pool2(e2)
        
        b = self.bottleneck(p2)
        
        d2 = self.dec2(torch.cat([self.up2(b), e2], dim=1))
        d1 = self.dec1(torch.cat([self.up1(d2), e1], dim=1))
        return self.out_conv(d1).squeeze(1)  # (B, H, W)

def binary_dice_loss(pred_logits: torch.Tensor, target_masks: torch.Tensor, smooth: float = 1.0) -> torch.Tensor:
    # Computes Soft Binary Dice Loss
    probs = torch.sigmoid(pred_logits)
    dims = (1, 2)
    intersection = torch.sum(probs * target_masks, dim=dims)
    cardinality = torch.sum(probs, dim=dims) + torch.sum(target_masks, dim=dims)
    dice = (2.0 * intersection + smooth) / (cardinality + smooth)
    return 1.0 - dice.mean()

def separate_instances(binary_mask: np.ndarray) -> np.ndarray:
    # Connected component labeling for instance segmentation post-processing
    labeled_array, num_features = ndimage.label(binary_mask > 0.5)
    return labeled_array

# Train CellUNet for 15 epochs with combined BCE + Dice Loss
set_seed(42)
cell_unet = CellUNet().to(device)
cell_opt = torch.optim.Adam(cell_unet.parameters(), lr=0.003)
bce_criterion = nn.BCEWithLogitsLoss()

epochs_seg = 15
best_val_dice = -1.0
best_unet_weights = copy.deepcopy(cell_unet.state_dict())

print(f"\n{'='*20} Training CellUNet ({epochs_seg} Epochs) {'='*20}")
for epoch in range(1, epochs_seg + 1):
    cell_unet.train()
    running_loss, running_dice, total_samples = 0.0, 0.0, 0
    
    pbar = tqdm(cell_train_loader, desc=f"[Segmentation] Epoch {epoch:2d}/{epochs_seg}", leave=False)
    for batch in pbar:
        imgs = batch["image"].to(device)
        masks = batch["semantic_mask"].to(device)
        
        cell_opt.zero_grad()
        pred_logits = cell_unet(imgs)
        loss = 0.5 * bce_criterion(pred_logits, masks) + 0.5 * binary_dice_loss(pred_logits, masks)
        loss.backward()
        cell_opt.step()
        
        b_size = imgs.size(0)
        running_loss += loss.item() * b_size
        probs = torch.sigmoid(pred_logits)
        inter = (probs * masks).sum(dim=(1, 2))
        card = probs.sum(dim=(1, 2)) + masks.sum(dim=(1, 2))
        batch_dice = (2.0 * inter / torch.clamp(card, min=1e-6)).sum().item()
        running_dice += batch_dice
        total_samples += b_size
        pbar.set_postfix(loss=f"{loss.item():.4f}", dice=f"{running_dice/total_samples*100:.1f}%")
        
    train_loss = running_loss / total_samples
    train_dice = (running_dice / total_samples) * 100.0
    
    # Validation step
    cell_unet.eval()
    val_loss, val_dice_sum, val_total = 0.0, 0.0, 0
    with torch.no_grad():
        for batch in cell_val_loader:
            imgs = batch["image"].to(device)
            masks = batch["semantic_mask"].to(device)
            pred_logits = cell_unet(imgs)
            v_loss = 0.5 * bce_criterion(pred_logits, masks) + 0.5 * binary_dice_loss(pred_logits, masks)
            
            b_size = imgs.size(0)
            val_loss += v_loss.item() * b_size
            probs = torch.sigmoid(pred_logits)
            inter = (probs * masks).sum(dim=(1, 2))
            card = probs.sum(dim=(1, 2)) + masks.sum(dim=(1, 2))
            val_dice_sum += (2.0 * inter / torch.clamp(card, min=1e-6)).sum().item()
            val_total += b_size
            
    val_loss /= val_total
    val_dice = (val_dice_sum / val_total) * 100.0
    
    print(f"Epoch {epoch:2d}/{epochs_seg} | Train Loss: {train_loss:.4f}, Dice: {train_dice:5.2f}% | Val Loss: {val_loss:.4f}, Val Dice: {val_dice:5.2f}%")
    
    if val_dice > best_val_dice:
        best_val_dice = val_dice
        best_unet_weights = copy.deepcopy(cell_unet.state_dict())

cell_unet.load_state_dict(best_unet_weights)
print(f"--> Restored Best Validation Checkpoint for CellUNet (Val Dice: {best_val_dice:.2f}%)")

# Test evaluation
cell_unet.eval()
test_dices, test_ious, test_preds = [], [], []
with torch.no_grad():
    for batch in cell_test_loader:
        imgs = batch["image"].to(device)
        masks = batch["semantic_mask"].to(device)
        pred_logits = cell_unet(imgs)
        pred_masks = (torch.sigmoid(pred_logits) > 0.5).float()
        
        for p_m, g_m in zip(pred_masks.cpu().numpy(), masks.cpu().numpy()):
            inter = np.logical_and(p_m == 1, g_m == 1).sum()
            union = np.logical_or(p_m == 1, g_m == 1).sum()
            iou = inter / max(1, union)
            dice = (2.0 * inter) / max(1, p_m.sum() + g_m.sum())
            test_ious.append(iou)
            test_dices.append(dice)
            test_preds.append(p_m)

mean_dice = np.mean(test_dices) * 100.0
mean_iou = np.mean(test_ious) * 100.0
print(f"\nCell Segmentation Test Results -> Test Dice Score: {mean_dice:.2f}%, Test IoU: {mean_iou:.2f}%")

# Visualize: Input Image | Semantic GT | Predicted Semantic | Predicted Instance
fig, axes = plt.subplots(2, 4, figsize=(16, 8))
for i in range(2):
    s = cell_test_dataset[i]
    p_mask = test_preds[i]
    p_inst = separate_instances(p_mask)
    
    axes[i, 0].imshow(s["image"].permute(1, 2, 0).numpy())
    axes[i, 0].set_title(f"Input Scene {i+1}", fontsize=11)
    axes[i, 0].axis('off')
    
    axes[i, 1].imshow(s["semantic_mask"].numpy(), cmap="gray")
    axes[i, 1].set_title(f"Semantic GT {i+1}", fontsize=11)
    axes[i, 1].axis('off')
    
    axes[i, 2].imshow(p_mask, cmap="gray")
    axes[i, 2].set_title(f"Predicted Semantic (Dice: {test_dices[i]*100:.1f}%)", fontsize=11)
    axes[i, 2].axis('off')
    
    axes[i, 3].imshow(p_inst, cmap="tab10", vmin=0, vmax=9)
    axes[i, 3].set_title(f"Predicted Instances ({p_inst.max()} Cells)", fontsize=11)
    axes[i, 3].axis('off')

plt.suptitle("Microscopy Cell Segmentation Results: Semantic vs Instance Segmentation", fontsize=14)
plt.tight_layout()
plt.show()
```
